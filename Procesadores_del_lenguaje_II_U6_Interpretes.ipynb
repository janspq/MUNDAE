## 🔷 SECCIÓN 1 — Conceptos Fundamentales de Intérpretes

Un **intérprete** es un programa que ejecuta directamente las instrucciones de un programa fuente, instrucción por instrucción, sin generar previamente un archivo ejecutable binario independiente. A diferencia del compilador —que traduce todo el programa a código máquina antes de correrlo— el intérprete **analiza y ejecuta al mismo tiempo**.

**Ciclo básico de un intérprete:**
```
1. OBTENER  → leer la siguiente instrucción
2. EVALUAR  → determinar su significado
3. EJECUTAR → realizar la acción correspondiente
4. Repetir hasta terminar o encontrar error
```

### 📝 Ejercicio 1 — Simulación del ciclo Obtener–Evaluar–Ejecutar

**Concepto:** El ciclo más básico de un intérprete: leer instrucciones de una lista y ejecutarlas una a una.

**Objetivo:** Comprender cómo un intérprete procesa instrucciones de forma secuencial.

````python
# EJERCICIO 1 — Ciclo Obtener-Evaluar-Ejecutar
# Simulamos un mini-intérprete que procesa instrucciones tipo 'PRINT <valor>'

# Lista de instrucciones del programa fuente
programa = [
    'PRINT Hola',
    'PRINT Mundo',
    'PRINT Procesadores del Lenguaje II',
]

print("=== INICIO DEL INTÉRPRETE ===")
print(f"Total de instrucciones: {len(programa)}\n")

# Ciclo principal del intérprete
for i, instruccion in enumerate(programa):
    # FASE 1: OBTENER — leer la instrucción actual
    print(f"[Paso {i+1}] OBTENER → '{instruccion}'")

    # FASE 2: EVALUAR — analizar qué operación es
    partes = instruccion.split(' ', 1)  # dividir en operacion + argumento
    operacion = partes[0]
    argumento = partes[1] if len(partes) > 1 else ''
    print(f"         EVALUAR → operación='{operacion}', argumento='{argumento}'")

    # FASE 3: EJECUTAR — realizar la acción
    if operacion == 'PRINT':
        print(f"         EJECUTAR → {argumento}")
    else:
        print(f"         EJECUTAR → ⚠️ Instrucción desconocida")
    print()

print("=== FIN DEL PROGRAMA ===")
````

**🔍 Análisis:**  
- Cada instrucción pasa por tres fases: obtener → evaluar → ejecutar.  
- El bucle `for` representa el **motor principal del intérprete**.  
- En un intérprete real, estas fases son más complejas (análisis léxico, sintáctico, etc.).

### 📝 Ejercicio 3 — Intérprete Puro vs Híbrido

**Concepto:** Los intérpretes puros ejecutan directamente el fuente; los híbridos compilan primero a bytecode y lo interpretan con una VM.  
**Objetivo:** Simular el flujo de ambos tipos.

````python
# EJERCICIO 3 — Intérprete Puro vs Sistema Híbrido (bytecode)

fuente = "x = 5 + 3"

print("=" * 50)
print("INTÉRPRETE PURO")
print("=" * 50)
print(f"Código fuente: '{fuente}'")
print("Proceso:")
print("  1. Leer código fuente directamente")
print("  2. Analizar léxica y sintácticamente EN TIEMPO REAL")
print("  3. Ejecutar instrucción por instrucción")
print("  → Resultado: se repite el análisis cada vez que se ejecuta el programa")

print()
print("=" * 50)
print("SISTEMA HÍBRIDO (compilación a bytecode)")
print("=" * 50)
print(f"Código fuente: '{fuente}'")

# Simular 'bytecode' como lista de instrucciones de bajo nivel
def compilar_a_bytecode(fuente):
    """Simulación simplificada de compilación a bytecode"""
    bytecode = []
    bytecode.append(("LOAD_CONST",  5))      # cargar constante 5
    bytecode.append(("LOAD_CONST",  3))      # cargar constante 3
    bytecode.append(("BINARY_ADD",  None))   # operación suma
    bytecode.append(("STORE_NAME",  'x'))    # guardar en variable x
    return bytecode

def ejecutar_bytecode(bytecode):
    """Mini máquina virtual que ejecuta el bytecode"""
    pila   = []   # pila de operandos
    entorno = {}  # tabla de variables
    for opcode, operando in bytecode:
        if opcode == "LOAD_CONST":
            pila.append(operando)
        elif opcode == "BINARY_ADD":
            b = pila.pop()
            a = pila.pop()
            pila.append(a + b)
        elif opcode == "STORE_NAME":
            entorno[operando] = pila.pop()
    return entorno

print("\n[FASE 1] Compilación a bytecode (una sola vez):")
bytecode = compilar_a_bytecode(fuente)
for idx, (op, arg) in enumerate(bytecode):
    print(f"  [{idx}] {op:<15} {arg}")

print("\n[FASE 2] Ejecución del bytecode por la VM:")
entorno = ejecutar_bytecode(bytecode)
print(f"  Variables en memoria: {entorno}")
print(f"  → x = {entorno['x']}  ✅")
````

**🔍 Análisis:**  
- En el sistema **híbrido**, el análisis léxico/sintáctico ocurre **una sola vez** al generar el bytecode.  
- La VM solo decodifica opcodes simples, reduciendo el overhead.  
- Python y Java usan exactamente este modelo (`.pyc` y `.class` respectivamente).

### 📝 Ejercicio 4 — REPL: Read-Eval-Print Loop

**Concepto:** El REPL es la interfaz interactiva característica de los intérpretes.  
**Objetivo:** Implementar un REPL básico que evalúe expresiones matemáticas.

````python
# EJERCICIO 4 — Simulación de un REPL
# En Colab/Jupyter no podemos usar input() en bucle,
# así que simulamos la sesión REPL con una lista de entradas.

def repl_simulado(entradas):
    """
    Simula un REPL básico.
    Acepta expresiones matemáticas y el comando 'salir'.
    """
    print("=== Mini REPL — Procesadores del Lenguaje II ===")
    print("Tipo de intérprete: PURO (evalúa expresión directamente)")
    print("Escribe 'salir' para terminar.\n")

    entorno = {}  # variables del usuario

    for entrada in entradas:
        print(f">>> {entrada}")

        if entrada.strip().lower() == 'salir':
            print("Hasta luego.")
            break

        try:
            # Evaluar la expresión usando el entorno de variables
            resultado = eval(entrada, {"__builtins__": {}}, entorno)
            if resultado is not None:
                print(resultado)
        except SyntaxError:
            print("⚠️ Error de sintaxis")
        except NameError as e:
            print(f"⚠️ Variable no definida: {e}")
        except Exception as e:
            print(f"⚠️ Error: {e}")
        print()

# Sesión REPL simulada
sesion = [
    "2 + 3",
    "10 * 4 - 5",
    "x = 7",        # asignación (no retorna valor)
    "x + 3",        # usa la variable definida
    "y",            # variable no definida → error
    "salir",
]

repl_simulado(sesion)
````

**🔍 Análisis:**  
- El ciclo **R**ead → **E**val → **P**rint → **L**oop (REPL) es la esencia de la interactividad de los intérpretes.  
- Lenguajes como Python, Ruby y JavaScript tienen REPL nativos.  
- El REPL mantiene un **entorno persistente** entre instrucciones (la variable `x` definida en un paso está disponible en el siguiente).

### 📝 Ejercicio 5 — Manejo Básico de Errores en Tiempo de Ejecución

**Concepto:** Los intérpretes detectan errores **durante la ejecución**, no antes. Esto es diferente a los compiladores que detectan muchos errores antes de ejecutar.  
**Objetivo:** Ver cómo un intérprete reporta errores con contexto.

````python
# EJERCICIO 5 — Detección de errores en tiempo de ejecución

def interpretar_programa(instrucciones):
    """
    Mini-intérprete que ejecuta asignaciones y muestra errores
    con número de línea (ventaja de los intérpretes: contexto del error).
    """
    entorno = {}
    print("--- Inicio de ejecución ---")

    for num_linea, instruccion in enumerate(instrucciones, start=1):
        print(f"  Línea {num_linea}: {instruccion}")
        try:
            exec(instruccion, {"__builtins__": {}}, entorno)
            # Mostrar el estado actual de variables
            vars_actuales = {k: v for k, v in entorno.items()}
            if vars_actuales:
                print(f"           Estado: {vars_actuales}")
        except ZeroDivisionError:
            print(f"           ❌ ERROR en línea {num_linea}: División por cero")
            print(f"           → El intérprete reporta la línea exacta del fallo")
        except NameError as e:
            print(f"           ❌ ERROR en línea {num_linea}: Variable no declarada — {e}")
        except TypeError as e:
            print(f"           ❌ ERROR en línea {num_linea}: Tipos incompatibles — {e}")

    print("--- Fin de ejecución ---")
    print(f"\nEstado final del programa: {entorno}")

# Programa con un error a mitad de la ejecución
programa_con_error = [
    "a = 10",
    "b = 0",
    "c = a / b",    # Error: división por cero (línea 3)
    "d = a + 5",    # Esta línea sí se ejecuta en algunos intérpretes
]

interpretar_programa(programa_con_error)
````

**🔍 Análisis:**  
- El intérprete ejecutó correctamente las líneas 1 y 2 antes de encontrar el error en la línea 3.  
- Un **compilador** hubiera rechazado el programa completo si hubiera detectado el error en análisis estático, pero `b=0` es un valor **dinámico**: solo el intérprete puede detectarlo en tiempo de ejecución.  
- El intérprete proporciona el **número de línea exacto** donde falló, lo cual es una ventaja en depuración.

## 🔷 SECCIÓN 2 — Análisis Léxico (Tokenización)

El **analizador léxico** (o *lexer/scanner*) es el primer componente de un intérprete. Su trabajo es leer el código fuente carácter a carácter y agruparlos en **tokens**: unidades significativas como palabras clave, identificadores, operadores y literales.

```
Entrada:  "x = 10 + y;"
Tokens:   [ID:'x']  [OP:'=']  [NUM:10]  [OP:'+']  [ID:'y']  [PUNCT:';']
```

### 📝 Ejercicio 6 — Tokenizador con Expresiones Regulares

**Concepto:** Un lexer usa expresiones regulares para reconocer patrones en el código fuente.  
**Objetivo:** Construir un tokenizador básico paso a paso.

````python
# EJERCICIO 6 — Tokenizador básico con expresiones regulares
import re

# Definición de tokens: (nombre_tipo, patron_regex)
PATRONES_TOKEN = [
    ('NUMERO',     r'\d+(\.\d+)?'),    # Enteros o flotantes: 42, 3.14
    ('CADENA',     r'"[^"]*"'),         # Cadenas: "hola"
    ('PALABRA_CL', r'\b(si|sino|mientras|retorna|var)\b'),  # Palabras clave
    ('IDENT',      r'[a-zA-Z_]\w*'),   # Identificadores: x, nombre, _val
    ('OP_ASIGN',   r'==|!=|<=|>='),    # Operadores relacionales compuestos
    ('OP',         r'[+\-*/=<>]'),     # Operadores simples
    ('LPAREN',     r'\('),
    ('RPAREN',     r'\)'),
    ('PUNTO_COMA', r';'),
    ('ESPACIO',    r'\s+'),             # Se ignora
]

def tokenizar(codigo):
    """Convierte una cadena de código en una lista de tokens."""
    tokens = []
    pos = 0
    while pos < len(codigo):
        coincidencia = None
        for tipo, patron in PATRONES_TOKEN:
            regex = re.compile(patron)
            coincidencia = regex.match(codigo, pos)
            if coincidencia:
                valor = coincidencia.group(0)
                if tipo != 'ESPACIO':       # ignorar espacios
                    tokens.append((tipo, valor))
                pos = coincidencia.end()
                break
        if not coincidencia:
            raise SyntaxError(f"Carácter inesperado: '{codigo[pos]}' en posición {pos}")
    return tokens

# Prueba con distintas expresiones
ejemplos = [
    'x = 10 + 3.5',
    'si (x >= 5) retorna x',
    'var nombre = "Juan";',
]

for codigo in ejemplos:
    print(f"\nCódigo: {codigo}")
    print("Tokens:")
    tokens = tokenizar(codigo)
    for tipo, valor in tokens:
        print(f"  ({tipo:12s}, {repr(valor)})")
````

**🔍 Análisis:**  
- El **orden** de los patrones importa: los operadores compuestos (`==`, `!=`) deben buscarse antes que los simples (`=`, `!`).  
- Los espacios se **ignoran** (no generan token), pero son necesarios como separadores.  
- El tokenizador reporta el **carácter y posición** exacta del error léxico.

### 📝 Ejercicio 7 — Clasificación de Tokens

**Concepto:** Los tokens se clasifican en categorías: números, identificadores, operadores, puntuación, etc.  
**Objetivo:** Contar y clasificar los tokens de un fragmento de código.

````python
# EJERCICIO 7 — Clasificación y conteo de tokens
from collections import Counter

# Reutilizamos la función tokenizar() del ejercicio anterior

programa = '''
var x = 10;
var y = 20;
var suma = x + y;
si (suma >= 25) retorna suma;
'''

tokens = tokenizar(programa.strip())

print("=" * 45)
print("ANÁLISIS LÉXICO COMPLETO")
print("=" * 45)
print(f"Total de tokens: {len(tokens)}\n")

# Contar por tipo
conteo_tipos = Counter(tipo for tipo, _ in tokens)

print("Distribución por tipo:")
for tipo, cantidad in conteo_tipos.most_common():
    barra = '█' * cantidad
    print(f"  {tipo:12s}: {barra} ({cantidad})")

print("\nLista de tokens generados:")
for i, (tipo, valor) in enumerate(tokens):
    print(f"  [{i:02d}] {tipo:12s} → {repr(valor)}")
````

**🔍 Análisis:**  
- El conteo de tipos de tokens revela la **densidad del código**: muchos `OP` indica código aritmético intensivo.  
- El índice de cada token (posición en la lista) permite al parser referirse a ellos eficientemente.  
- Esta estructura de tokens es exactamente lo que el analizador sintáctico recibirá como entrada.

### 📝 Ejercicio 8 — Detección de Errores Léxicos

**Concepto:** El analizador léxico debe detectar caracteres o secuencias no permitidos por el lenguaje.  
**Objetivo:** Implementar manejo de errores léxicos con mensajes descriptivos.

````python
# EJERCICIO 8 — Errores léxicos con contexto

def tokenizar_con_errores(codigo):
    """Tokenizador que colecta TODOS los errores léxicos antes de fallar."""
    tokens = []
    errores = []
    pos = 0
    while pos < len(codigo):
        coincidencia = None
        for tipo, patron in PATRONES_TOKEN:
            regex = re.compile(patron)
            coincidencia = regex.match(codigo, pos)
            if coincidencia:
                valor = coincidencia.group(0)
                if tipo != 'ESPACIO':
                    tokens.append((tipo, valor, pos))
                pos = coincidencia.end()
                break
        if not coincidencia:
            char_malo = codigo[pos]
            errores.append(f"  Error léxico en posición {pos}: carácter ilegal '{char_malo}'")
            pos += 1  # continuar analizando
    return tokens, errores

# Código con errores léxicos: @ y # no son válidos en nuestro lenguaje
codigo_con_errores = 'x = 10 @ 3 + #5'

print(f"Código de entrada: '{codigo_con_errores}'\n")
tokens, errores = tokenizar_con_errores(codigo_con_errores)

print("Tokens válidos encontrados:")
for tipo, valor, pos in tokens:
    print(f"  pos={pos:02d} ({tipo:12s}, {repr(valor)})")

print()
if errores:
    print(f"⚠️  Se encontraron {len(errores)} error(es) léxico(s):")
    for e in errores:
        print(e)
else:
    print("✅ Sin errores léxicos")
````

**🔍 Análisis:**  
- Un buen lexer **no se detiene** al primer error; recopila todos los errores para dar al programador una vista completa.  
- La posición exacta del error permite al IDE resaltarla en rojo.  
- Este comportamiento de "recuperación de errores" es estándar en intérpretes profesionales como CPython.

### 📝 Ejercicio 9 — Reconocimiento de Palabras Clave vs Identificadores

**Concepto:** La distinción entre palabras clave (`if`, `while`) e identificadores (`x`, `nombre`) es fundamental en el lexer.  
**Objetivo:** Implementar esa distinción explícitamente.

````python
# EJERCICIO 9 — Palabras clave vs identificadores

PALABRAS_CLAVE = {'si', 'sino', 'mientras', 'para', 'retorna', 'var',
                  'verdadero', 'falso', 'nulo', 'funcion'}

def clasificar_identificador(lexema):
    """Determina si un lexema es palabra clave o identificador."""
    if lexema in PALABRAS_CLAVE:
        return 'PALABRA_CLAVE'
    return 'IDENTIFICADOR'

# Lista de lexemas a clasificar
lexemas = ['x', 'si', 'contador', 'mientras', 'si2', 'retorna',
           'nombre_usuario', 'funcion', 'verdadero', 'miVariable']

print(f"{'Lexema':<20} {'Tipo':<15} {'¿Reservado?'}")
print("-" * 48)
for lexema in lexemas:
    tipo = clasificar_identificador(lexema)
    reservado = "✅ Sí" if tipo == 'PALABRA_CLAVE' else "❌ No"
    print(f"{lexema:<20} {tipo:<15} {reservado}")

print(f"\nTotal palabras clave definidas: {len(PALABRAS_CLAVE)}")
print("Las palabras clave NO pueden usarse como nombres de variables.")
````

**🔍 Análisis:**  
- `si2` es un **identificador válido** aunque contiene 'si': las palabras clave son exactas.  
- En la mayoría de lenguajes, las palabras clave son **sensibles a mayúsculas** (`if` ≠ `IF`).  
- Este es el primer paso que diferencia el significado de los tokens en el análisis semántico.

### 📝 Ejercicio 10 — Lexer con Números Enteros y Flotantes

**Concepto:** Distinguir entre tipos de literales numéricos es importante para el análisis semántico.  
**Objetivo:** Tokenizar y categorizar literales numéricos.

````python
# EJERCICIO 10 — Distinción de literales numéricos
import re

def analizar_literal_numerico(texto):
    """Clasifica si un literal numérico es entero o flotante."""
    patron_float = re.compile(r'^-?\d+\.\d+$')
    patron_int   = re.compile(r'^-?\d+$')

    if patron_float.match(texto):
        return ('FLOAT', float(texto))
    elif patron_int.match(texto):
        return ('INT', int(texto))
    else:
        return ('DESCONOCIDO', texto)

# Ejemplos a clasificar
numeros = ['42', '3.14', '-7', '-2.5', '0', '100.0', '1e5', 'abc']

print(f"{'Texto':<12} {'Tipo':<12} {'Valor Python':<15} {'Tamaño (bytes)'}")
print("-" * 56)
import sys
for num in numeros:
    tipo, valor = analizar_literal_numerico(num)
    tam = sys.getsizeof(valor) if tipo != 'DESCONOCIDO' else '-'
    print(f"{num:<12} {tipo:<12} {str(valor):<15} {tam}")

print("\n💡 Los enteros y flotantes ocupan diferente memoria y tienen operaciones distintas.")
````

**🔍 Análisis:**  
- La distinción `INT` vs `FLOAT` es crucial para la **verificación de tipos** en el análisis semántico.  
- El lexer almacena el **valor ya convertido** (no solo el texto) para eficiencia en la evaluación posterior.  
- `1e5` no matchea nuestros patrones simples: en un lexer real se añadiría soporte para notación científica.

## 🔷 SECCIÓN 3 — Árbol de Sintaxis Abstracta (AST)

El **AST** (Abstract Syntax Tree) es la representación interna del programa después del análisis sintáctico. Captura la **estructura jerárquica** del programa, eliminando detalles sintácticos irrelevantes (como paréntesis y puntos y coma) y resaltando las relaciones semánticas entre las partes del código.

```
Expresión:  10 + y * 2
AST:
        (+)
       /   \
     10     (*)
           /   \
          y     2
```

### 📝 Ejercicio 11 — Nodos del AST (Clases Base)

**Concepto:** El AST se compone de nodos, cada uno con un tipo y posibles hijos.  
**Objetivo:** Definir las clases de nodos para un lenguaje simple.

````python
# EJERCICIO 11 — Definición de nodos del AST

class NodoAST:
    """Clase base para todos los nodos del AST."""
    def __repr__(self):
        return f"{self.__class__.__name__}()"

class NodoNumero(NodoAST):
    """Nodo hoja: representa un literal numérico."""
    def __init__(self, valor):
        self.valor = valor
    def __repr__(self):
        return f"Numero({self.valor})"

class NodoIdentificador(NodoAST):
    """Nodo hoja: representa una variable."""
    def __init__(self, nombre):
        self.nombre = nombre
    def __repr__(self):
        return f"Var({self.nombre})"

class NodoOperacionBinaria(NodoAST):
    """Nodo interno: representa una operación con dos operandos."""
    def __init__(self, operador, izquierda, derecha):
        self.operador  = operador    # '+', '-', '*', '/'
        self.izquierda = izquierda   # NodoAST
        self.derecha   = derecha     # NodoAST
    def __repr__(self):
        return f"BinOp({self.operador}, {self.izquierda}, {self.derecha})"

class NodoAsignacion(NodoAST):
    """Nodo: representa variable = expresion."""
    def __init__(self, variable, expresion):
        self.variable  = variable   # NodoIdentificador
        self.expresion = expresion  # NodoAST
    def __repr__(self):
        return f"Asignar({self.variable}, {self.expresion})"

# Construir manualmente el AST para:  x = 10 + y * 2
ast_manual = NodoAsignacion(
    variable   = NodoIdentificador('x'),
    expresion  = NodoOperacionBinaria(
        operador  = '+',
        izquierda = NodoNumero(10),
        derecha   = NodoOperacionBinaria(
            operador  = '*',
            izquierda = NodoIdentificador('y'),
            derecha   = NodoNumero(2)
        )
    )
)

print("Representación del AST para: x = 10 + y * 2")
print(ast_manual)
print("\nAcceso a partes del árbol:")
print(f"  Variable asignada : {ast_manual.variable}")
print(f"  Operador raíz exp : {ast_manual.expresion.operador}")
print(f"  Operando izquierdo: {ast_manual.expresion.izquierda}")
print(f"  Operando derecho  : {ast_manual.expresion.derecha}")
````

**🔍 Análisis:**  
- Los **nodos hoja** (`NodoNumero`, `NodoIdentificador`) no tienen hijos.  
- Los **nodos internos** (`NodoOperacionBinaria`) tienen hijos que son también nodos AST.  
- Esta estructura jerárquica respeta la **precedencia de operadores** (la multiplicación `y*2` es un subárbol dentro de la suma).

**Objetivo:** Imprimir el AST en formato de árbol con indentación.

````python
# EJERCICIO 12 — Visualización del AST con indentación

def imprimir_ast(nodo, nivel=0, prefijo=""):
    """Imprime el AST como árbol visual."""
    sangria = "  " * nivel
    if isinstance(nodo, NodoNumero):
        print(f"{sangria}{prefijo}📄 Número: {nodo.valor}")
    elif isinstance(nodo, NodoIdentificador):
        print(f"{sangria}{prefijo}🔤 Var: {nodo.nombre}")
    elif isinstance(nodo, NodoOperacionBinaria):
        print(f"{sangria}{prefijo}➕ OperBin: '{nodo.operador}'")
        imprimir_ast(nodo.izquierda, nivel+1, "├─ izq: ")
        imprimir_ast(nodo.derecha,   nivel+1, "└─ der: ")
    elif isinstance(nodo, NodoAsignacion):
        print(f"{sangria}{prefijo}🏷️  Asignacion")
        imprimir_ast(nodo.variable,  nivel+1, "├─ var: ")
        imprimir_ast(nodo.expresion, nivel+1, "└─ exp: ")

print("AST para: x = 10 + y * 2")
print("=" * 35)
imprimir_ast(ast_manual)

# Segundo ejemplo: a = (b + 3) * (c - 1)
ast2 = NodoAsignacion(
    NodoIdentificador('a'),
    NodoOperacionBinaria('*',
        NodoOperacionBinaria('+', NodoIdentificador('b'), NodoNumero(3)),
        NodoOperacionBinaria('-', NodoIdentificador('c'), NodoNumero(1))
    )
)

print("\nAST para: a = (b + 3) * (c - 1)")
print("=" * 35)
imprimir_ast(ast2)
````

### 📝 Ejercicio 13 — AST para Sentencias de Control

**Concepto:** Las sentencias de control (`if`, `while`) también se representan como nodos del AST.  
**Objetivo:** Definir y construir nodos para sentencias de control.

````python
# EJERCICIO 13 — Nodos AST para sentencias de control

class NodoCadena(NodoAST):
    """Literal de cadena de texto."""
    def __init__(self, valor):
        self.valor = valor
    def __repr__(self):
        return f"Cadena({repr(self.valor)})"

class NodoBloque(NodoAST):
    """Secuencia de sentencias."""
    def __init__(self, sentencias):
        self.sentencias = sentencias  # lista de NodoAST
    def __repr__(self):
        return f"Bloque({len(self.sentencias)} sentencias)"

class NodoSi(NodoAST):
    """Sentencia if-else."""
    def __init__(self, condicion, bloque_si, bloque_sino=None):
        self.condicion   = condicion
        self.bloque_si   = bloque_si
        self.bloque_sino = bloque_sino

class NodoMientras(NodoAST):
    """Sentencia while."""
    def __init__(self, condicion, cuerpo):
        self.condicion = condicion
        self.cuerpo    = cuerpo

class NodoImprimir(NodoAST):
    """Instrucción print."""
    def __init__(self, expresion):
        self.expresion = expresion

# Construir AST para:
#   si (x > 5):
#       imprimir("grande")
#   sino:
#       imprimir("pequeño")

ast_si = NodoSi(
    condicion   = NodoOperacionBinaria('>', NodoIdentificador('x'), NodoNumero(5)),
    bloque_si   = NodoBloque([NodoImprimir(NodoCadena("grande"))]),
    bloque_sino = NodoBloque([NodoImprimir(NodoCadena("pequeño"))])
)

def describir_ast(nodo, nivel=0):
    """Descripción recursiva del AST."""
    ind = "  " * nivel
    if isinstance(nodo, NodoSi):
        print(f"{ind}[SI]")
        print(f"{ind}  condición:")
        describir_ast(nodo.condicion, nivel+2)
        print(f"{ind}  bloque_si:")
        describir_ast(nodo.bloque_si, nivel+2)
        if nodo.bloque_sino:
            print(f"{ind}  bloque_sino:")
            describir_ast(nodo.bloque_sino, nivel+2)
    elif isinstance(nodo, NodoBloque):
        for s in nodo.sentencias:
            describir_ast(s, nivel)
    elif isinstance(nodo, NodoImprimir):
        print(f"{ind}[IMPRIMIR] {nodo.expresion}")
    elif isinstance(nodo, NodoOperacionBinaria):
        print(f"{ind}[OP '{nodo.operador}'] {nodo.izquierda}  vs  {nodo.derecha}")
    else:
        print(f"{ind}{nodo}")

print("AST para: si (x > 5): imprimir('grande') sino: imprimir('pequeño')")
print("=" * 60)
describir_ast(ast_si)
````

**🔍 Análisis:**  
- La sentencia `if-else` tiene **tres hijos**: condición, bloque-entonces, bloque-sino (opcional).  
- El `NodoBloque` agrupa sentencias en secuencia dentro de `{}` o indentación.  
- La `condicion` es ella misma un subárbol de expresión (`x > 5`).

### 📝 Ejercicio 14 — Conteo de Nodos y Profundidad del AST

**Concepto:** Métricas sobre el AST pueden indicar la complejidad del programa.  
**Objetivo:** Calcular el número de nodos y la profundidad máxima del árbol.

````python
# EJERCICIO 14 — Métricas del AST

def contar_nodos(nodo):
    """Cuenta el total de nodos en el AST."""
    if isinstance(nodo, (NodoNumero, NodoIdentificador, NodoCadena)):
        return 1
    elif isinstance(nodo, NodoOperacionBinaria):
        return 1 + contar_nodos(nodo.izquierda) + contar_nodos(nodo.derecha)
    elif isinstance(nodo, NodoAsignacion):
        return 1 + contar_nodos(nodo.variable) + contar_nodos(nodo.expresion)
    elif isinstance(nodo, NodoSi):
        total = 1 + contar_nodos(nodo.condicion) + contar_nodos(nodo.bloque_si)
        if nodo.bloque_sino:
            total += contar_nodos(nodo.bloque_sino)
        return total
    elif isinstance(nodo, NodoBloque):
        return 1 + sum(contar_nodos(s) for s in nodo.sentencias)
    elif isinstance(nodo, NodoImprimir):
        return 1 + contar_nodos(nodo.expresion)
    return 1

def profundidad_ast(nodo):
    """Calcula la profundidad máxima del árbol."""
    if isinstance(nodo, (NodoNumero, NodoIdentificador, NodoCadena)):
        return 1
    elif isinstance(nodo, NodoOperacionBinaria):
        return 1 + max(profundidad_ast(nodo.izquierda), profundidad_ast(nodo.derecha))
    elif isinstance(nodo, NodoAsignacion):
        return 1 + max(profundidad_ast(nodo.variable), profundidad_ast(nodo.expresion))
    elif isinstance(nodo, NodoSi):
        profundidades = [profundidad_ast(nodo.condicion), profundidad_ast(nodo.bloque_si)]
        if nodo.bloque_sino:
            profundidades.append(profundidad_ast(nodo.bloque_sino))
        return 1 + max(profundidades)
    elif isinstance(nodo, NodoBloque):
        if not nodo.sentencias:
            return 1
        return 1 + max(profundidad_ast(s) for s in nodo.sentencias)
    elif isinstance(nodo, NodoImprimir):
        return 1 + profundidad_ast(nodo.expresion)
    return 1

# Analizar los ASTs de ejercicios anteriores
arboles = [
    ("x = 10 + y * 2",                ast_manual),
    ("a = (b+3) * (c-1)",             ast2),
    ("si (x>5) imprimir...",           ast_si),
]

print(f"{'Expresión':<35} {'Nodos':<8} {'Profundidad'}")
print("-" * 55)
for nombre, arbol in arboles:
    n = contar_nodos(arbol)
    p = profundidad_ast(arbol)
    print(f"{nombre:<35} {n:<8} {p}")
````

**🔍 Análisis:**  
- Más nodos = programa más complejo.  
- Mayor profundidad = expresiones más anidadas = posible **desbordamiento de pila** en la recursión del evaluador.  
- Estas métricas se usan en análisis de complejidad de compiladores e intérpretes.

### 📝 Ejercicio 15 — Parser Recursivo Descendente Simple

**Concepto:** El parser construye el AST a partir de los tokens usando gramáticas recursivas.  
**Objetivo:** Implementar un parser básico para expresiones aritméticas.

````python
# EJERCICIO 15 — Parser recursivo descendente para expresiones
# Gramática soportada:
#   expr   → termino (('+' | '-') termino)*
#   termino → factor  (('*' | '/') factor)*
#   factor  → NUMERO | IDENT | '(' expr ')'

class Parser:
    def __init__(self, tokens):
        self.tokens = tokens
        self.pos    = 0

    def token_actual(self):
        if self.pos < len(self.tokens):
            return self.tokens[self.pos]
        return ('EOF', '')

    def consumir(self, tipo_esperado=None):
        tok = self.token_actual()
        if tipo_esperado and tok[0] != tipo_esperado:
            raise SyntaxError(f"Se esperaba {tipo_esperado}, se obtuvo {tok}")
        self.pos += 1
        return tok

    def parsear_factor(self):
        tok = self.token_actual()
        if tok[0] == 'NUMERO':
            self.consumir('NUMERO')
            # Convertir a int o float según tenga punto decimal
            val = float(tok[1]) if '.' in tok[1] else int(tok[1])
            return NodoNumero(val)
        elif tok[0] == 'IDENT':
            self.consumir('IDENT')
            return NodoIdentificador(tok[1])
        elif tok[0] == 'LPAREN':
            self.consumir('LPAREN')
            nodo = self.parsear_expr()
            self.consumir('RPAREN')
            return nodo
        else:
            raise SyntaxError(f"Token inesperado: {tok}")

    def parsear_termino(self):
        nodo = self.parsear_factor()
        while self.token_actual()[1] in ('*', '/'):
            op  = self.consumir('OP')[1]
            der = self.parsear_factor()
            nodo = NodoOperacionBinaria(op, nodo, der)
        return nodo

    def parsear_expr(self):
        nodo = self.parsear_termino()
        while self.token_actual()[1] in ('+', '-'):
            op  = self.consumir('OP')[1]
            der = self.parsear_termino()
            nodo = NodoOperacionBinaria(op, nodo, der)
        return nodo

# Probar con distintas expresiones
expresiones = [
    '3 + 4',
    '2 + 3 * 4',         # precedencia: * antes que +
    '(2 + 3) * 4',       # paréntesis alteran precedencia
    'x + 10',
]

for expr in expresiones:
    tokens_expr = tokenizar(expr)
    # Filtrar solo los tipos que el parser conoce
    tokens_filtrados = [(t,v) for t,v in tokens_expr
                        if t in ('NUMERO','IDENT','OP','LPAREN','RPAREN')]
    parser = Parser(tokens_filtrados)
    ast_resultado = parser.parsear_expr()
    print(f"  '{expr}'  →  {ast_resultado}")
````

**🔍 Análisis:**  
- `parsear_expr` maneja `+` y `-`; `parsear_termino` maneja `*` y `/`. Esto implementa la **precedencia de operadores**.  
- `2 + 3 * 4` genera `BinOp(+, 2, BinOp(*, 3, 4))`, no `BinOp(*, BinOp(+, 2, 3), 4)`.  
- Este es el patrón **recursivo descendente predictivo**, el más usado en intérpretes caseros.

## 🔷 SECCIÓN 4 — Evaluación de Expresiones

La **evaluación** es el corazón del intérprete: recorre el AST recursivamente y ejecuta cada operación. Para expresiones se usa recorrido **postorden** (evaluar hijos antes que el padre).

### 📝 Ejercicio 16 — Evaluador de Expresiones Aritméticas

**Concepto:** El evaluador recorre el AST y calcula el valor de cada nodo.  
**Objetivo:** Implementar un evaluador recursivo completo.

````python
# EJERCICIO 16 — Evaluador de expresiones sobre AST

class Evaluador:
    def __init__(self, entorno=None):
        # entorno: diccionario de variables {nombre: valor}
        self.entorno = entorno if entorno is not None else {}

    def evaluar(self, nodo):
        """Evalúa un nodo AST y retorna su valor."""
        # Caso base: literales
        if isinstance(nodo, NodoNumero):
            return nodo.valor

        # Caso base: cadenas de texto
        elif isinstance(nodo, NodoCadena):
            return nodo.valor

        # Caso base: variables
        elif isinstance(nodo, NodoIdentificador):
            if nodo.nombre not in self.entorno:
                raise NameError(f"Variable '{nodo.nombre}' no definida")
            return self.entorno[nodo.nombre]

        # Caso recursivo: operaciones binarias
        elif isinstance(nodo, NodoOperacionBinaria):
            # Primero evaluar AMBOS hijos (postorden)
            val_izq = self.evaluar(nodo.izquierda)
            val_der = self.evaluar(nodo.derecha)

            # Luego aplicar el operador
            if   nodo.operador == '+':
                return val_izq + val_der
            elif nodo.operador == '-':
                return val_izq - val_der
            elif nodo.operador == '*':
                return val_izq * val_der
            elif nodo.operador == '/':
                if val_der == 0:
                    raise ZeroDivisionError("División por cero")
                return val_izq / val_der
            elif nodo.operador == '>':
                return val_izq > val_der
            elif nodo.operador == '<':
                return val_izq < val_der
            elif nodo.operador == '>=':
                return val_izq >= val_der
            elif nodo.operador == '<=':
                return val_izq <= val_der
            elif nodo.operador == '==':
                return val_izq == val_der
            elif nodo.operador == '!=':
                return val_izq != val_der
            else:
                raise ValueError(f"Operador desconocido: '{nodo.operador}'")

        raise TypeError(f"Tipo de nodo no reconocido: {type(nodo)}")

# Crear evaluador con variables
entorno_vars = {'x': 10, 'y': 3}
evaluador = Evaluador(entorno_vars)

# ASTs de prueba
casos = [
    ("5 + 3",         NodoOperacionBinaria('+', NodoNumero(5), NodoNumero(3))),
    ("x * 2",         NodoOperacionBinaria('*', NodoIdentificador('x'), NodoNumero(2))),
    ("x + y * 2",     NodoOperacionBinaria('+', NodoIdentificador('x'),
                         NodoOperacionBinaria('*', NodoIdentificador('y'), NodoNumero(2)))),
    ("x > 5",         NodoOperacionBinaria('>', NodoIdentificador('x'), NodoNumero(5))),
    ("10 / 0",        NodoOperacionBinaria('/', NodoNumero(10), NodoNumero(0))),
]

print(f"Entorno: {entorno_vars}\n")
for desc, nodo_prueba in casos:
    try:
        resultado = evaluador.evaluar(nodo_prueba)
        print(f"  {desc:<20} → {resultado}")
    except Exception as e:
        print(f"  {desc:<20} → ❌ {type(e).__name__}: {e}")
````

**🔍 Análisis:**  
- La evaluación de `NodoOperacionBinaria` evalúa **primero los hijos** antes de aplicar el operador (postorden).  
- El acceso a variables requiere consultar el **entorno** (tabla de símbolos).  
- Los errores como división por cero se detectan **en tiempo de ejecución**, no antes.

### 📝 Ejercicio 17 — Evaluación con Tipos Dinámicos

**Concepto:** Los intérpretes dinámicos permiten que el tipo de una variable cambie en tiempo de ejecución.  
**Objetivo:** Demostrar la verificación de tipos en tiempo de ejecución.

````python
# EJERCICIO 17 — Tipado dinámico y verificación de tipos en ejecución

def evaluar_con_tipos(op, val_izq, val_der):
    """
    Evaluador con verificación de tipos dinámica.
    Implementa coerción de tipos donde es seguro.
    """
    tipo_izq = type(val_izq).__name__
    tipo_der = type(val_der).__name__

    print(f"  Operación: {val_izq} ({tipo_izq}) {op} {val_der} ({tipo_der})")

    if op == '+':
        # Regla: int + int = int; float + cualquier_num = float; str + str = str
        if isinstance(val_izq, str) and isinstance(val_der, str):
            resultado = val_izq + val_der
            print(f"  → Concatenación de cadenas: {repr(resultado)}")
        elif isinstance(val_izq, (int, float)) and isinstance(val_der, (int, float)):
            resultado = val_izq + val_der
            print(f"  → Suma numérica: {resultado}")
        else:
            raise TypeError(f"No se puede sumar '{tipo_izq}' con '{tipo_der}'")
        return resultado
    elif op == '*':
        if isinstance(val_izq, str) and isinstance(val_der, int):
            resultado = val_izq * val_der
            print(f"  → Repetición de cadena: {repr(resultado)}")
            return resultado
        elif isinstance(val_izq, (int,float)) and isinstance(val_der, (int,float)):
            resultado = val_izq * val_der
            print(f"  → Multiplicación: {resultado}")
            return resultado
        else:
            raise TypeError(f"No se puede multiplicar '{tipo_izq}' por '{tipo_der}'")

# Casos de evaluación con tipos
print("=== Evaluación con tipos dinámicos ===\n")
casos_tipos = [
    ('+', 5,       3),
    ('+', 2.5,     3),
    ('+', "hola",  " mundo"),
    ('*', "abc",   3),
    ('+', 5,       "texto"),   # Error esperado
]

for op, a, b in casos_tipos:
    try:
        evaluar_con_tipos(op, a, b)
    except TypeError as e:
        print(f"  ❌ TypeError: {e}")
    print()
````

**🔍 Análisis:**  
- En Python (intérprete dinámico), `5 + 3` y `"hola" + " mundo"` ambos usan `+` pero con semánticas diferentes.  
- La **verificación de tipos** ocurre **en tiempo de ejecución**, no antes.  
- Esto es diferente a C++ (compilado estático) donde mezclar tipos genera error en compilación.

### 📝 Ejercicio 18 — Evaluación de Expresiones Lógicas

**Concepto:** Las expresiones lógicas devuelven booleanos y se usan en condiciones de control.  
**Objetivo:** Evaluar expresiones booleanas compuestas.

````python
# EJERCICIO 18 — Evaluación de expresiones lógicas

class NodoOperacionLogica(NodoAST):
    """Operaciones AND, OR, NOT sobre expresiones booleanas."""
    def __init__(self, operador, izquierda, derecha=None):
        self.operador  = operador   # 'Y', 'O', 'NO'
        self.izquierda = izquierda
        self.derecha   = derecha    # None para NOT

class EvaluadorLogico(Evaluador):
    """Extiende el evaluador para manejar operaciones lógicas."""

    def evaluar(self, nodo):
        if isinstance(nodo, NodoOperacionLogica):
            if nodo.operador == 'NO':
                val = self.evaluar(nodo.izquierda)
                return not val
            elif nodo.operador == 'Y':
                # Evaluación de cortocircuito: si izq es False, no evaluar der
                val_izq = self.evaluar(nodo.izquierda)
                print(f"    AND: izquierda={val_izq}", end="")
                if not val_izq:
                    print(" → cortocircuito, no evaluar derecha")
                    return False
                val_der = self.evaluar(nodo.derecha)
                print(f", derecha={val_der}")
                return val_der
            elif nodo.operador == 'O':
                val_izq = self.evaluar(nodo.izquierda)
                print(f"    OR:  izquierda={val_izq}", end="")
                if val_izq:
                    print(" → cortocircuito, no evaluar derecha")
                    return True
                val_der = self.evaluar(nodo.derecha)
                print(f", derecha={val_der}")
                return val_der
        return super().evaluar(nodo)

# Variables de prueba
env = {'edad': 20, 'tiene_permiso': True, 'es_miembro': False}
ev  = EvaluadorLogico(env)

# Expresión: (edad > 18) AND tiene_permiso
expr1 = NodoOperacionLogica('Y',
    NodoOperacionBinaria('>', NodoIdentificador('edad'), NodoNumero(18)),
    NodoIdentificador('tiene_permiso')
)

# Expresión: es_miembro OR (edad > 18)
expr2 = NodoOperacionLogica('O',
    NodoIdentificador('es_miembro'),
    NodoOperacionBinaria('>', NodoIdentificador('edad'), NodoNumero(18))
)

# Expresión: NOT es_miembro
expr3 = NodoOperacionLogica('NO', NodoIdentificador('es_miembro'))

print(f"Entorno: {env}\n")
print("(edad > 18) Y tiene_permiso:")
print(f"  Resultado: {ev.evaluar(expr1)}\n")
print("es_miembro O (edad > 18):")
print(f"  Resultado: {ev.evaluar(expr2)}\n")
print("NO es_miembro:")
print(f"  Resultado: {ev.evaluar(expr3)}")
````

**🔍 Análisis:**  
- La **evaluación de cortocircuito** (short-circuit) es una optimización importante: si `A AND B` ya sabe que A es False, no evalúa B (ahorra tiempo y evita errores).  
- Este comportamiento es estándar en Python (`and`, `or`) y la mayoría de lenguajes modernos.  
- `NOT` es un operador **unario**: solo tiene un operando (el hijo izquierdo).

### 📝 Ejercicio 19 — Evaluación de Llamadas a Funciones

**Concepto:** Las llamadas a función crean un nuevo ámbito (frame), ejecutan el cuerpo y retornan un valor.  
**Objetivo:** Simular la evaluación de una llamada a función con paso de parámetros.

````python
# EJERCICIO 19 — Evaluación de llamadas a función

class NodoFuncion(NodoAST):
    """Definición de función."""
    def __init__(self, nombre, parametros, cuerpo):
        self.nombre     = nombre       # str
        self.parametros = parametros   # lista de str
        self.cuerpo     = cuerpo       # NodoAST (bloque)

class NodoLlamada(NodoAST):
    """Llamada a función."""
    def __init__(self, nombre_func, argumentos):
        self.nombre_func = nombre_func   # str
        self.argumentos  = argumentos    # lista de NodoAST

class NodoRetorna(NodoAST):
    """Sentencia return."""
    def __init__(self, expresion):
        self.expresion = expresion

class ExcepcionRetorno(Exception):
    """Usada para propagar el valor de retorno por el árbol."""
    def __init__(self, valor):
        self.valor = valor

def ejecutar_funcion(definicion, args_evaluados, evaluador):
    """
    Ejecuta una función:
    1. Crea nuevo ámbito local
    2. Asocia parámetros ↔ argumentos
    3. Ejecuta el cuerpo
    4. Retorna el valor de retorno
    """
    print(f"  → Llamando '{definicion.nombre}' con args: {args_evaluados}")

    # Crear nuevo entorno local (copia del global + parámetros)
    entorno_local = dict(evaluador.entorno)  # heredar globales
    for param, valor in zip(definicion.parametros, args_evaluados):
        entorno_local[param] = valor
        print(f"     Parámetro '{param}' = {valor}")

    # Ejecutar el cuerpo con el entorno local
    evaluador_local = Evaluador(entorno_local)
    return evaluador_local.evaluar(definicion.cuerpo)

# Definir función: sumar(a, b) → a + b
func_sumar = NodoFuncion(
    nombre     = 'sumar',
    parametros = ['a', 'b'],
    cuerpo     = NodoOperacionBinaria('+', NodoIdentificador('a'), NodoIdentificador('b'))
)

# Definir función: cuadrado(x) → x * x
func_cuadrado = NodoFuncion(
    nombre     = 'cuadrado',
    parametros = ['x'],
    cuerpo     = NodoOperacionBinaria('*', NodoIdentificador('x'), NodoIdentificador('x'))
)

# Evaluador base
ev_base = Evaluador({'global_val': 100})

print("=== Evaluación de funciones ===\n")
# Llamar sumar(3, 7)
resultado1 = ejecutar_funcion(func_sumar,    [3, 7],  ev_base)
print(f"  sumar(3, 7) = {resultado1}\n")

# Llamar cuadrado(5)
resultado2 = ejecutar_funcion(func_cuadrado, [5],     ev_base)
print(f"  cuadrado(5) = {resultado2}\n")

# Llamar sumar(cuadrado(3), 1) — composición
val_cuad = ejecutar_funcion(func_cuadrado, [3], ev_base)
resultado3 = ejecutar_funcion(func_sumar, [val_cuad, 1], ev_base)
print(f"  sumar(cuadrado(3), 1) = {resultado3}")
````

**🔍 Análisis:**  
- Al llamar una función se **crea un nuevo entorno local** que hereda el global pero tiene sus propias variables.  
- Los **parámetros** se vinculan con los **argumentos** evaluados antes de entrar al cuerpo.  
- La composición `sumar(cuadrado(3), 1)` requiere evaluar el argumento interno primero: esto es recursión en la evaluación.

### 📝 Ejercicio 20 — Recorrido Post-orden del AST

**Concepto:** El recorrido post-orden evalúa primero los hijos y luego el nodo padre, ideal para expresiones.  
**Objetivo:** Visualizar el orden de evaluación de un AST en postorden.

````python
# EJERCICIO 20 — Visualización del recorrido post-orden

def evaluar_con_traza(nodo, entorno, nivel=0):
    """
    Evalúa el AST mostrando el orden exacto de operaciones.
    Recorrido POST-ORDEN: hijos antes que padre.
    """
    ind = "  " * nivel

    if isinstance(nodo, NodoNumero):
        print(f"{ind}↓ Evaluar Número({nodo.valor}) → {nodo.valor}")
        return nodo.valor

    elif isinstance(nodo, NodoIdentificador):
        val = entorno.get(nodo.nombre, None)
        print(f"{ind}↓ Evaluar Var({nodo.nombre}) → {val}")
        return val

    elif isinstance(nodo, NodoOperacionBinaria):
        print(f"{ind}↓ Iniciar OperBin('{nodo.operador}')")

        # POST-ORDEN: evaluar izquierda primero
        val_izq = evaluar_con_traza(nodo.izquierda, entorno, nivel+1)
        # Luego derecha
        val_der = evaluar_con_traza(nodo.derecha,   entorno, nivel+1)

        # Por último, el padre
        resultado = eval(f"{val_izq} {nodo.operador} {val_der}")
        print(f"{ind}↑ OperBin('{nodo.operador}'): {val_izq} {nodo.operador} {val_der} = {resultado}")
        return resultado

# Expresión: (2 + 3) * (4 - 1)
ast_postorden = NodoOperacionBinaria('*',
    NodoOperacionBinaria('+', NodoNumero(2), NodoNumero(3)),
    NodoOperacionBinaria('-', NodoNumero(4), NodoNumero(1))
)

print("Expresión: (2 + 3) * (4 - 1)")
print("Orden de evaluación (post-orden — ↓ entrar, ↑ resultado):")
print("-" * 50)
resultado = evaluar_con_traza(ast_postorden, {})
print("-" * 50)
print(f"Resultado final: {resultado}")
````

**🔍 Análisis:**  
- La traza muestra que `2+3` y `4-1` se evalúan **antes** que la multiplicación: esto es post-orden.  
- Las flechas `↓` (entrar al nodo) y `↑` (obtener resultado) simulan la pila de llamadas recursivas.  
- Este orden garantiza que los operandos siempre estén disponibles cuando se necesitan.

## 🔷 SECCIÓN 5 — Tabla de Símbolos y Ámbitos

La **tabla de símbolos** almacena información de todos los identificadores del programa. En un intérprete es dinámica: almacena no solo tipos sino también los **valores actuales**. La **gestión de ámbitos** (scoping) permite que el mismo nombre tenga diferentes valores en diferentes partes del programa.

### 📝 Ejercicio 21 — Tabla de Símbolos Básica

**Concepto:** La tabla de símbolos es un diccionario de identificadores a sus atributos.  
**Objetivo:** Implementar una tabla de símbolos con operaciones CRUD.

````python
# EJERCICIO 21 — Tabla de Símbolos básica

class TablaSimbolos:
    """Tabla de símbolos para un intérprete: almacena nombre → {tipo, valor}"""

    def __init__(self, nombre_ambito='global'):
        self.nombre_ambito = nombre_ambito
        self._tabla = {}  # {nombre: {'tipo': str, 'valor': any}}

    def declarar(self, nombre, tipo, valor=None):
        """Declara un nuevo símbolo en la tabla."""
        if nombre in self._tabla:
            raise NameError(f"'{nombre}' ya fue declarado en este ámbito")
        self._tabla[nombre] = {'tipo': tipo, 'valor': valor}
        print(f"  [DECLARAR] '{nombre}': tipo={tipo}, valor={valor}")

    def asignar(self, nombre, valor):
        """Actualiza el valor de una variable existente."""
        if nombre not in self._tabla:
            raise NameError(f"'{nombre}' no declarado")
        tipo_actual = self._tabla[nombre]['tipo']
        self._tabla[nombre]['valor'] = valor
        print(f"  [ASIGNAR]  '{nombre}' = {valor}  (tipo: {tipo_actual})")

    def buscar(self, nombre):
        """Retorna el registro de un símbolo."""
        if nombre not in self._tabla:
            raise NameError(f"'{nombre}' no definido")
        return self._tabla[nombre]

    def mostrar(self):
        """Muestra el contenido de la tabla."""
        print(f"\n  Tabla de Símbolos [{self.nombre_ambito}]")
        print(f"  {'Nombre':<15} {'Tipo':<10} {'Valor'}")
        print("  " + "-" * 35)
        for nombre, info in self._tabla.items():
            print(f"  {nombre:<15} {info['tipo']:<10} {info['valor']}")

# Demostración
ts = TablaSimbolos('global')
print("=== Operaciones sobre la Tabla de Símbolos ===\n")

ts.declarar('x',       'entero',  0)
ts.declarar('nombre',  'cadena',  'Ana')
ts.declarar('precio',  'flotante', 9.99)

ts.asignar('x', 42)
ts.asignar('precio', 12.5)

ts.mostrar()

print("\nBúsqueda de 'nombre':", ts.buscar('nombre'))

try:
    ts.buscar('z')   # variable no declarada
except NameError as e:
    print(f"\n❌ Error: {e}")
````

**🔍 Análisis:**  
- La tabla almacena tanto el **tipo** como el **valor actual** de cada variable.  
- `declarar` y `asignar` son operaciones distintas: declarar crea la entrada; asignar la actualiza.  
- La búsqueda de una variable no declarada genera un `NameError`, el mismo error que lanza Python en tiempo de ejecución.

### 📝 Ejercicio 22 — Gestión de Ámbitos (Scoping)

**Concepto:** El ámbito define dónde es visible una variable. Se implementa como una **pila de tablas de símbolos**.  
**Objetivo:** Implementar ámbitos anidados con búsqueda en cadena.

````python
# EJERCICIO 22 — Pila de ámbitos (Entorno con cadena de ámbitos)

class Entorno:
    """
    Implementa un entorno con ámbitos anidados.
    Cada ámbito es un diccionario; los ámbitos forman una cadena.
    """
    def __init__(self, padre=None, nombre='global'):
        self._vars  = {}      # variables de este ámbito
        self.padre  = padre   # ámbito padre (None = ámbito global)
        self.nombre = nombre

    def definir(self, nombre, valor):
        """Define una variable en el ámbito ACTUAL."""
        self._vars[nombre] = valor

    def obtener(self, nombre):
        """Busca la variable empezando en el ámbito actual y subiendo."""
        if nombre in self._vars:
            return self._vars[nombre]
        elif self.padre is not None:
            return self.padre.obtener(nombre)  # buscar en padre
        else:
            raise NameError(f"Variable '{nombre}' no definida en ningún ámbito")

    def asignar(self, nombre, valor):
        """Actualiza la variable en el ámbito donde fue definida."""
        if nombre in self._vars:
            self._vars[nombre] = valor
        elif self.padre is not None:
            self.padre.asignar(nombre, valor)
        else:
            raise NameError(f"Variable '{nombre}' no definida")

    def cadena_ambitos(self):
        """Muestra la cadena completa de ámbitos."""
        resultado = []
        actual = self
        while actual is not None:
            resultado.append(f"[{actual.nombre}]: {actual._vars}")
            actual = actual.padre
        return " → ".join(resultado)

# Demostración de ámbitos anidados
print("=== Demostración de Ámbitos Anidados ===\n")

# Ámbito global
global_env = Entorno(nombre='global')
global_env.definir('x', 10)
global_env.definir('y', 20)
print(f"Global:   {global_env._vars}")

# Ámbito de función (hijo del global)
func_env = Entorno(padre=global_env, nombre='funcion_f')
func_env.definir('x', 99)    # oculta la 'x' global
func_env.definir('z', 5)
print(f"Función:  {func_env._vars}")

# Ámbito de bloque if (hijo de función)
bloque_env = Entorno(padre=func_env, nombre='bloque_if')
bloque_env.definir('temp', 42)
print(f"Bloque:   {bloque_env._vars}")

print("\n--- Búsqueda de variables desde el ámbito del bloque ---")
print(f"  x     = {bloque_env.obtener('x')}   ← de función (oculta global)")
print(f"  y     = {bloque_env.obtener('y')}   ← del global (no en func ni bloque)")
print(f"  z     = {bloque_env.obtener('z')}    ← de función")
print(f"  temp  = {bloque_env.obtener('temp')} ← del bloque")

print(f"\nCadena de ámbitos:\n  {bloque_env.cadena_ambitos()}")
````

**🔍 Análisis:**  
- La variable `x` tiene valor `99` en la función (oculta la `x=10` global): esto es **variable shadowing**.  
- La variable `y` no está en la función ni en el bloque, pero se encuentra **subiendo la cadena** hasta el global.  
- Al salir del bloque `if`, `temp` desaparece (se descarta el `bloque_env`).

### 📝 Ejercicio 23 — Creación y Destrucción de Ámbitos

**Concepto:** Cuando se llama a una función, se **crea** un nuevo ámbito; cuando retorna, se **destruye**.  
**Objetivo:** Simular el ciclo de vida de los ámbitos con una pila.

````python
# EJERCICIO 23 — Ciclo de vida de ámbitos (pila de marcos)

class PilaMarcos:
    """
    Simula la pila de marcos de activación de un intérprete.
    Cada llamada a función empuja un nuevo marco;
    cada retorno lo saca.
    """
    def __init__(self):
        self._pila = [{'nombre': 'global', 'vars': {}}]

    def entrar_funcion(self, nombre_func, params):
        """Crea un nuevo marco para la función."""
        marco = {'nombre': nombre_func, 'vars': dict(params)}
        self._pila.append(marco)
        self._mostrar_pila(f"ENTRAR {nombre_func}")

    def salir_funcion(self):
        """Destruye el marco actual al retornar."""
        if len(self._pila) > 1:
            marco = self._pila.pop()
            self._mostrar_pila(f"SALIR  {marco['nombre']}")

    def definir_var(self, nombre, valor):
        """Define variable en el marco actual."""
        self._pila[-1]['vars'][nombre] = valor

    def _mostrar_pila(self, evento):
        """Visualiza la pila de marcos."""
        print(f"\n[{evento}]")
        for i, marco in enumerate(reversed(self._pila)):
            nivel = len(self._pila) - 1 - i
            ind = '  ' * nivel
            tag = "← actual" if i == 0 else ""
            print(f"  {'▓' * (nivel+1)} [{marco['nombre']}] vars={marco['vars']} {tag}")

# Simular: main() llama a calcular(a=5), que llama a doble(n=5)
pila = PilaMarcos()
pila.definir_var('resultado_global', 0)

print("Simulación de llamadas anidadas:")
print("  main() → calcular(a=5) → doble(n=5) → retorna → retorna → retorna")

pila.entrar_funcion('calcular', {'a': 5})
pila.definir_var('b', 10)
pila.entrar_funcion('doble', {'n': 5})
pila.definir_var('resultado', 10)
pila.salir_funcion()   # sale doble
pila.salir_funcion()   # sale calcular

print("\nEstado final (solo queda el ámbito global):")
pila._mostrar_pila("FIN")
````

**🔍 Análisis:**  
- La pila crece al entrar a cada función y decrece al salir: esto es el **Call Stack** que vemos en los stack traces de errores.  
- Las variables de `doble` (como `resultado`) desaparecen cuando se hace pop del marco.  
- Un **StackOverflow** ocurre cuando la pila crece demasiado (recursión infinita).

### 📝 Ejercicio 24 — Ámbito Léxico vs Ámbito Dinámico

**Concepto:** El **ámbito léxico** (estático) resuelve variables según la estructura del código; el **ámbito dinámico** según la cadena de llamadas en ejecución.  
**Objetivo:** Comparar ambos mecanismos con un ejemplo.

````python
# EJERCICIO 24 — Ámbito léxico vs ámbito dinámico

print("=" * 55)
print("ÁMBITO LÉXICO (estático) — usado por Python, JS, C")
print("=" * 55)
print("Las funciones 'capturan' el entorno donde fueron DEFINIDAS.\n")

# En Python: ámbito léxico
x_global = "global"   # variable en ámbito global

def funcion_externa():
    x_local = "externa"   # variable en ámbito de funcion_externa

    def funcion_interna():
        # Busca x en su ámbito (no existe),
        # luego en el ámbito LÉXICO padre (funcion_externa)
        print(f"  Dentro de funcion_interna, x = '{x_local}'")
        print(f"  (resuelve x del ámbito donde fue DEFINIDA, no donde es LLAMADA)")

    return funcion_interna  # retorna la función (closure)

def llamador():
    x_llamador = "llamador"
    fi = funcion_externa()   # obtener la función interna
    fi()                     # llamarla desde ámbito diferente

llamador()

print("\n" + "=" * 55)
print("ÁMBITO DINÁMICO — simulación")
print("=" * 55)
print("La función busca x en la cadena de llamadas en EJECUCIÓN.\n")

# Simular ámbito dinámico con pila de variables
pila_dinamica = [{'x': 'global'}]

def buscar_dinamico(nombre):
    for marco in reversed(pila_dinamica):
        if nombre in marco:
            return marco[nombre]
    raise NameError(f"'{nombre}' no encontrado")

def funcion_interna_din():
    print(f"  Ámbito dinámico — x = '{buscar_dinamico('x')}'")

# Llamar desde contexto donde x='llamador'
pila_dinamica.append({'x': 'llamador'})
print("Llamada desde contexto con x='llamador':")
funcion_interna_din()
pila_dinamica.pop()

# Llamar desde contexto sin x local
pila_dinamica.append({'y': 'otro'})
print("Llamada desde contexto sin x local (busca en global):")
funcion_interna_din()
pila_dinamica.pop()
````

**🔍 Análisis:**  
- **Léxico**: la función `funcion_interna` siempre ve `x='externa'` (donde fue definida), sin importar desde dónde se llame.  
- **Dinámico**: la función busca `x` en el stack de llamadas activo en el momento de ejecución.  
- Python, JavaScript y la mayoría de lenguajes modernos usan **ámbito léxico**: es más predecible y fácil de razonar.

### 📝 Ejercicio 25 — Closures: Funciones que Capturan su Entorno

**Concepto:** Una *closure* es una función que "recuerda" el entorno donde fue definida, incluso después de que ese entorno ya no esté activo.  
**Objetivo:** Implementar y demostrar closures.

````python
# EJERCICIO 25 — Closures (funciones que capturan su entorno)

def crear_contador(inicio=0):
    """
    Crea y retorna una función que incrementa un contador.
    El contador vive en el entorno de crear_contador,
    pero la función interna lo recuerda (closure).
    """
    conteo = [inicio]  # lista para permitir mutación en closure de Python

    def incrementar(paso=1):
        conteo[0] += paso
        return conteo[0]

    return incrementar

def crear_multiplicador(factor):
    """Retorna una función que multiplica por un factor fijo."""
    def multiplicar(numero):
        return numero * factor   # 'factor' capturado del entorno exterior
    return multiplicar

print("=== Demostración de Closures ===\n")

# Dos contadores independientes, cada uno con su propio entorno capturado
contador_a = crear_contador(0)
contador_b = crear_contador(100)

print("Contadores independientes:")
print(f"  contador_a: {contador_a()} → {contador_a()} → {contador_a(5)}")
print(f"  contador_b: {contador_b()} → {contador_b()} → {contador_b()}")
print("  (cada closure mantiene su propio estado)")

print()
# Multiplicadores con diferentes factores capturados
doble  = crear_multiplicador(2)
triple = crear_multiplicador(3)

print("Multiplicadores generados por closure:")
for n in [1, 5, 10]:
    print(f"  doble({n})={doble(n)}   triple({n})={triple(n)}")

print()
print("En un intérprete, las closures requieren que el entorno")
print("se PRESERVE en el heap aunque la función padre ya retornó.")
````

**🔍 Análisis:**  
- `contador_a` y `contador_b` tienen **entornos capturados independientes**: modificar uno no afecta al otro.  
- El intérprete debe mantener el entorno capturado en el **heap** (no en la pila) para que persista después del retorno de `crear_contador`.  
- Este es uno de los casos donde la gestión del entorno en intérpretes es más compleja que en compiladores.

### 📝 Ejercicio 26 — Tabla de Símbolos con Funciones

**Concepto:** Las funciones también son símbolos y se almacenan en la tabla junto con las variables.  
**Objetivo:** Extender la tabla de símbolos para almacenar funciones.

````python
# EJERCICIO 26 — Tabla de Símbolos con variables y funciones

class TablaSimbolosCompleta:
    """Tabla de símbolos que maneja tanto variables como funciones."""

    def __init__(self):
        self._simbolos = {}

    def definir_variable(self, nombre, tipo, valor):
        self._simbolos[nombre] = {
            'categoria': 'variable',
            'tipo': tipo,
            'valor': valor
        }

    def definir_funcion(self, nombre, parametros, tipo_retorno):
        self._simbolos[nombre] = {
            'categoria':     'funcion',
            'parametros':    parametros,   # [(nombre, tipo), ...]
            'tipo_retorno':  tipo_retorno,
        }

    def buscar(self, nombre):
        if nombre not in self._simbolos:
            raise NameError(f"'{nombre}' no definido")
        return self._simbolos[nombre]

    def es_funcion(self, nombre):
        info = self.buscar(nombre)
        return info['categoria'] == 'funcion'

    def mostrar(self):
        print(f"\n  {'Nombre':<15} {'Categoría':<12} {'Detalle'}")
        print("  " + "-" * 55)
        for nombre, info in self._simbolos.items():
            if info['categoria'] == 'variable':
                detalle = f"tipo={info['tipo']}, valor={info['valor']}"
            else:
                params = ', '.join(f"{n}:{t}" for n, t in info['parametros'])
                detalle = f"params=({params}), retorna={info['tipo_retorno']}"
            print(f"  {nombre:<15} {info['categoria']:<12} {detalle}")

# Construir tabla para un programa típico
ts = TablaSimbolosCompleta()

# Variables globales
ts.definir_variable('pi',      'flotante', 3.14159)
ts.definir_variable('maximo',  'entero',   100)
ts.definir_variable('nombre',  'cadena',   'mundo')

# Funciones
ts.definir_funcion('calcular_area', [('radio', 'flotante')], 'flotante')
ts.definir_funcion('saludar',       [('nombre', 'cadena')],  'cadena')
ts.definir_funcion('sumar',         [('a','entero'),('b','entero')], 'entero')

ts.mostrar()

print("\n¿'sumar' es función?",     ts.es_funcion('sumar'))
print("¿'pi' es función?",          ts.es_funcion('pi'))
````

**🔍 Análisis:**  
- En la tabla de símbolos, **variables y funciones coexisten** con distintas categorías.  
- Antes de llamar `sumar(1, 2)`, el intérprete consulta la tabla para verificar que es una función y cuántos parámetros acepta.  
- Esta información es fundamental para el **análisis semántico** (verificación de tipos y aridad).

### 📝 Ejercicio 27 — Ejecución de Asignaciones

**Concepto:** La asignación evalúa la expresión derecha y actualiza la tabla de símbolos.  
**Objetivo:** Implementar la ejecución de sentencias de asignación.

````python
# EJERCICIO 27 — Ejecución de sentencias de asignación

def ejecutar_asignacion(nodo_asign, entorno, evaluador):
    """
    Ejecuta: variable = expresion
    Pasos:
      1. Evaluar la expresión del lado derecho
      2. Buscar (o crear) la variable en la tabla
      3. Actualizar el valor
    """
    nombre_var = nodo_asign.variable.nombre

    # Paso 1: evaluar expresión
    valor = evaluador.evaluar(nodo_asign.expresion)
    print(f"  [ASIGNAR] Evaluar expresión → {valor}")

    # Paso 2 y 3: actualizar entorno
    valor_anterior = entorno.get(nombre_var, '<no definida>')
    entorno[nombre_var] = valor

    print(f"  [ASIGNAR] '{nombre_var}': {valor_anterior} → {valor}")
    print(f"  [ESTADO]  entorno = {entorno}")

# Entorno inicial
env = {'y': 5}
ev  = Evaluador(env)

# Sentencias de asignación en forma de AST
sentencias = [
    # x = 10
    NodoAsignacion(NodoIdentificador('x'), NodoNumero(10)),
    # x = x + y   (x=10, y=5 → x=15)
    NodoAsignacion(NodoIdentificador('x'),
        NodoOperacionBinaria('+', NodoIdentificador('x'), NodoIdentificador('y'))),
    # z = x * 2   (x=15 → z=30)
    NodoAsignacion(NodoIdentificador('z'),
        NodoOperacionBinaria('*', NodoIdentificador('x'), NodoNumero(2))),
]

print(f"Entorno inicial: {env}\n")
for i, s in enumerate(sentencias, 1):
    print(f"Sentencia {i}: {s}")
    ejecutar_asignacion(s, env, ev)
    print()
````

**🔍 Análisis:**  
- La expresión del lado derecho se evalúa **primero**, antes de actualizar la variable.  
- `x = x + y` funciona porque al evaluar `x + y`, se lee el valor **actual** de `x` (no el nuevo).  
- El evaluador actualiza el entorno **compartido**: por eso `z = x * 2` ya ve el nuevo valor de `x=15`.

### 📝 Ejercicio 28 — Ejecución de Sentencias Condicionales (if-else)

**Concepto:** El `if-else` evalúa la condición y ejecuta solo uno de los dos bloques.  
**Objetivo:** Implementar la ejecución de sentencias condicionales.

````python
# EJERCICIO 28 — Ejecución de sentencias condicionales

def ejecutar_bloque(bloque, entorno, evaluador_cls=Evaluador):
    """Ejecuta todas las sentencias de un bloque secuencialmente."""
    ev = evaluador_cls(entorno)
    resultados = []
    for sentencia in bloque.sentencias:
        if isinstance(sentencia, NodoAsignacion):
            val = ev.evaluar(sentencia.expresion)
            entorno[sentencia.variable.nombre] = val
            resultados.append(f"    {sentencia.variable.nombre} = {val}")
        elif isinstance(sentencia, NodoImprimir):
            val = ev.evaluar(sentencia.expresion)
            resultados.append(f"    IMPRIMIR: {val}")
    return resultados

def ejecutar_si(nodo_si, entorno):
    """
    Ejecuta una sentencia if-else:
      1. Evaluar la condición
      2. Si verdadera: ejecutar bloque_si
      3. Si falsa:     ejecutar bloque_sino (si existe)
    """
    ev = Evaluador(entorno)

    # Paso 1: evaluar condición
    condicion_val = ev.evaluar(nodo_si.condicion)
    print(f"  Condición evaluada: {condicion_val}")

    # Pasos 2 o 3: ejecutar bloque apropiado
    if condicion_val:
        print("  → Ejecutando bloque SI:")
        resultados = ejecutar_bloque(nodo_si.bloque_si, entorno)
    else:
        if nodo_si.bloque_sino:
            print("  → Ejecutando bloque SINO:")
            resultados = ejecutar_bloque(nodo_si.bloque_sino, entorno)
        else:
            print("  → Sin bloque SINO, no se ejecuta nada")
            resultados = []

    for r in resultados:
        print(r)

# Caso 1: nota = 75, si nota >= 60: aprobado sino: reprobado
env1 = {'nota': 75}
ast_si_nota = NodoSi(
    NodoOperacionBinaria('>=', NodoIdentificador('nota'), NodoNumero(60)),
    NodoBloque([NodoImprimir(NodoCadena("Aprobado ✅"))]),
    NodoBloque([NodoImprimir(NodoCadena("Reprobado ❌"))])
)

print(f"\nEscenario 1: nota = {env1['nota']}")
ejecutar_si(ast_si_nota, env1)

# Caso 2: nota = 45
env2 = {'nota': 45}
print(f"\nEscenario 2: nota = {env2['nota']}")
ejecutar_si(ast_si_nota, env2)
````

**🔍 Análisis:**  
- Solo **uno** de los dos bloques se ejecuta, nunca ambos. Esto es fundamental para la semántica del `if-else`.  
- La condición `nota >= 60` es evaluada como expresión antes de decidir cuál bloque ejecutar.  
- Si la condición es `False` y no hay `sino`, el intérprete simplemente pasa a la siguiente sentencia.

### 📝 Ejercicio 29 — Ejecución de Bucles While

**Concepto:** El bucle `while` evalúa la condición repetidamente y ejecuta el cuerpo mientras sea verdadera.  
**Objetivo:** Implementar la ejecución de bucles while con seguimiento de iteraciones.

````python
# EJERCICIO 29 — Ejecución de bucle while

def ejecutar_mientras(nodo_mientras, entorno, max_iter=20):
    """
    Ejecuta un bucle while:
      1. Evaluar condición
      2. Si True: ejecutar cuerpo → ir a paso 1
      3. Si False: salir del bucle
    max_iter: límite de seguridad para evitar bucles infinitos
    """
    ev        = Evaluador(entorno)
    iteracion = 0

    print(f"  Estado inicial: {entorno}")

    while True:
        # Paso 1: evaluar condición
        condicion = ev.evaluar(nodo_mientras.condicion)

        if not condicion:
            print(f"  Condición falsa en iteración {iteracion}: saliendo del bucle")
            break

        if iteracion >= max_iter:
            print(f"  ⚠️ Límite de {max_iter} iteraciones alcanzado (protección)")
            break

        # Paso 2: ejecutar cuerpo
        for sentencia in nodo_mientras.cuerpo.sentencias:
            if isinstance(sentencia, NodoAsignacion):
                val = ev.evaluar(sentencia.expresion)
                nombre = sentencia.variable.nombre
                entorno[nombre] = val
                ev.entorno[nombre] = val   # sincronizar

        iteracion += 1
        print(f"  Iter {iteracion:2d}: estado={dict(entorno)}")

    return entorno

# Programa: contador = 0; mientras contador < 5: contador = contador + 1
env = {'contador': 0}
ast_while = NodoMientras(
    condicion = NodoOperacionBinaria('<', NodoIdentificador('contador'), NodoNumero(5)),
    cuerpo    = NodoBloque([
        NodoAsignacion(
            NodoIdentificador('contador'),
            NodoOperacionBinaria('+', NodoIdentificador('contador'), NodoNumero(1))
        )
    ])
)

print("Programa: contador = 0; mientras (contador < 5): contador += 1")
print("-" * 55)
entorno_final = ejecutar_mientras(ast_while, env)
print(f"\n  Estado final: {entorno_final}")
````

**🔍 Análisis:**  
- El bucle evalúa la condición **antes de cada iteración**, incluyendo la primera.  
- El `max_iter` es una protección contra bucles infinitos (error de programación común).  
- El entorno se actualiza en cada iteración: el cambio de `contador` en la iteración N es visible en la condición de la iteración N+1.

### 📝 Ejercicio 30 — Sentencias break y continue

**Concepto:** `break` y `continue` alteran el flujo normal del bucle usando excepciones internas del intérprete.  
**Objetivo:** Implementar `break` y `continue` usando excepciones de señalización.

````python
# EJERCICIO 30 — Sentencias break y continue con excepciones de señal

class SeñalBreak(Exception):    pass  # señal para salir del bucle
class SeñalContinue(Exception): pass  # señal para siguiente iteración

def ejecutar_con_control(programa_lista, entorno_inicial):
    """
    Ejecuta una lista de instrucciones con soporte para break/continue.
    Las instrucciones son strings simples para facilitar la demo.
    """
    env = dict(entorno_inicial)
    i   = 0

    print(f"  Entorno inicial: {env}")
    print()

    while i < 10:  # límite del bucle
        try:
            env['i'] = i
            instruccion = programa_lista[i % len(programa_lista)]

            print(f"  Iter i={i}: ejecutar '{instruccion}'")

            if instruccion == 'break':
                raise SeñalBreak()
            elif instruccion == 'continue':
                i += 1
                raise SeñalContinue()
            elif instruccion.startswith('print'):
                valor = eval(instruccion.split('(')[1].rstrip(')'), {}, env)
                print(f"    → OUTPUT: {valor}")

        except SeñalContinue:
            print("    → continue: saltando al siguiente ciclo")
            continue
        except SeñalBreak:
            print("    → break: saliendo del bucle")
            break

        i += 1

    print(f"\n  Fin del bucle. i={i}")

# Programa 1: imprimir pares, saltar impares (continue)
print("Programa 1: imprimir hasta i==3 y luego break")
print("-" * 45)
prog1 = ['print(i)', 'break' if True else 'x']

# Simular directamente para mayor claridad
print("  Bucle: imprimir i mientras i<5, break cuando i==3")
for i in range(10):
    if i == 3:
        print(f"    i={i}: BREAK — salir del bucle")
        break
    print(f"    i={i}: imprimir {i}")

print()
print("Programa 2: saltar múltiplos de 2 (continue)")
print("-" * 45)
print("  Bucle: imprimir i si i%2!=0, continuar si i%2==0")
for i in range(8):
    if i % 2 == 0:
        print(f"    i={i}: CONTINUE — saltar este valor")
        continue
    print(f"    i={i}: imprimir {i}")
````

**🔍 Análisis:**  
- `break` y `continue` interrumpen el flujo **normal** del recorrido del AST.  
- En un intérprete, se implementan típicamente con **excepciones internas** que propagan la señal hacia arriba en la pila de recursión hasta el manejador del bucle.  
- Esta técnica (usar excepciones como señales de flujo de control) es exactamente cómo CPython implementa `break`, `continue` y `return`.

### 📝 Ejercicio 31 — Ejecución de Funciones Recursivas

**Concepto:** La recursión en funciones requiere que el intérprete cree un nuevo frame por cada llamada.  
**Objetivo:** Ejecutar una función recursiva y ver la pila de llamadas.

````python
# EJERCICIO 31 — Función recursiva: factorial

# Representamos las funciones como callables Python
# para demostrar los conceptos sin un parser completo

call_stack = []   # pila de llamadas simulada

def factorial_interpretado(n, nivel=0):
    """
    Simula cómo un intérprete ejecutaría la función factorial recursiva.
    Muestra la creación/destrucción de frames.
    """
    ind = "  " * nivel

    # Crear nuevo frame
    frame = {'funcion': 'factorial', 'n': n}
    call_stack.append(frame)
    print(f"{ind}→ ENTRAR factorial(n={n})  [Pila: {len(call_stack)} frames]")

    # Caso base
    if n == 0 or n == 1:
        resultado = 1
        print(f"{ind}  Caso base: retorna {resultado}")
    else:
        # Caso recursivo: n * factorial(n-1)
        sub_resultado = factorial_interpretado(n - 1, nivel + 1)
        resultado     = n * sub_resultado
        print(f"{ind}  Combinar: {n} * {sub_resultado} = {resultado}")

    # Destruir frame al retornar
    call_stack.pop()
    print(f"{ind}← SALIR  factorial(n={n}) = {resultado}  [Pila: {len(call_stack)} frames]")
    return resultado

print("Cálculo de factorial(5) con trazado de pila:")
print("=" * 55)
resultado = factorial_interpretado(5)
print("=" * 55)
print(f"\nResultado: factorial(5) = {resultado}")
print(f"Pila al final: {len(call_stack)} frames (vacía = correcto)")
````

**🔍 Análisis:**  
- Cada llamada recursiva **añade un frame** a la pila; cada retorno **elimina un frame**.  
- Al final, la pila queda **vacía**, lo que confirma que todos los frames se manejaron correctamente.  
- Si `n` fuera muy grande (ej. `factorial(10000)`), la pila crecería tanto que causaría un **RecursionError** (StackOverflow).

### 📝 Ejercicio 32 — Sentencia return y propagación del valor

**Concepto:** La sentencia `return` debe detener la ejecución del cuerpo y propagar el valor por la pila.  
**Objetivo:** Implementar `return` con propagación mediante excepciones internas.

````python
# EJERCICIO 32 — Implementación de sentencia return

class ValorRetorno(Exception):
    """Excepción usada para propagar el valor de return por el AST."""
    def __init__(self, valor):
        self.valor = valor

def ejecutar_cuerpo_funcion(sentencias, entorno):
    """
    Ejecuta las sentencias del cuerpo de una función.
    Si se encuentra un 'return', lanza ValorRetorno.
    """
    ev = Evaluador(entorno)
    for sentencia in sentencias:
        if isinstance(sentencia, NodoAsignacion):
            val = ev.evaluar(sentencia.expresion)
            entorno[sentencia.variable.nombre] = val
            ev.entorno[sentencia.variable.nombre] = val
            print(f"    Asignar {sentencia.variable.nombre} = {val}")
        elif isinstance(sentencia, NodoRetorna):
            valor_retorno = ev.evaluar(sentencia.expresion)
            print(f"    RETURN {valor_retorno} → propagar por la pila")
            raise ValorRetorno(valor_retorno)
        elif isinstance(sentencia, NodoImprimir):
            val = ev.evaluar(sentencia.expresion)
            print(f"    PRINT: {val}")

def llamar_funcion_con_return(nombre, params, cuerpo_sents, args):
    """Llama a una función y captura el valor de return."""
    entorno_local = dict(zip([p for p in params], args))
    print(f"  Entrando a '{nombre}' con {entorno_local}")
    try:
        ejecutar_cuerpo_funcion(cuerpo_sents, entorno_local)
        print(f"  Función terminó sin return → retorna None")
        return None
    except ValorRetorno as vr:
        print(f"  Función '{nombre}' retorna: {vr.valor}")
        return vr.valor

# Función: maximo(a, b) → si a > b: return a; sino: return b
cuerpo_maximo = [
    NodoSi(
        NodoOperacionBinaria('>', NodoIdentificador('a'), NodoIdentificador('b')),
        NodoBloque([NodoRetorna(NodoIdentificador('a'))]),
        NodoBloque([NodoRetorna(NodoIdentificador('b'))])
    )
]

# Adaptar para la ejecución simplificada
print("Función maximo(a, b):")
for a, b in [(7, 3), (2, 8), (5, 5)]:
    print(f"\n  maximo({a}, {b}):")
    env = {'a': a, 'b': b}
    # Simular el if-else con return
    try:
        if a > b:
            print(f"    Condición a>b = True → return a")
            raise ValorRetorno(a)
        else:
            print(f"    Condición a>b = False → return b")
            raise ValorRetorno(b)
    except ValorRetorno as vr:
        print(f"    Resultado: {vr.valor}")
````

**🔍 Análisis:**  
- El `return` usa una **excepción interna** (`ValorRetorno`) para interrumpir el recorrido del AST y propagar el valor.  
- Sin este mecanismo, el `return` dentro de un bloque anidado (como dentro de un `if`) no podría terminar la función.  
- Esta técnica es usada en CPython, MicroPython y muchos intérpretes de código abierto.

### 📝 Ejercicio 33 — Mini-Intérprete Integrado

**Concepto:** Un intérprete completo une lexer + parser + evaluador + tabla de símbolos.  
**Objetivo:** Construir y ejecutar un mini-intérprete end-to-end.

````python
# EJERCICIO 33 — Mini-intérprete integrado (Lexer → Parser → Evaluador)

class MiniInterprete:
    """
    Intérprete completo para expresiones de asignación simples.
    Soporta: números, identificadores, +, -, *, /, paréntesis.
    """

    def __init__(self):
        self.entorno = {}  # tabla de símbolos global

    def ejecutar(self, codigo):
        """Pipeline completo: código fuente → resultado."""
        print(f"\n[INPUT]   {repr(codigo)}")

        # FASE 1: Análisis léxico
        tokens = tokenizar(codigo)
        tokens_filtrados = [(t,v) for t,v in tokens
                            if t in ('NUMERO','IDENT','OP','LPAREN','RPAREN','OP_ASIGN')]
        print(f"[LÉXICO]  {[(t,v) for t,v in tokens_filtrados]}")

        # FASE 2: Análisis sintáctico (construir AST)
        # Detectar si es asignación: IDENT '=' expr
        if (len(tokens_filtrados) >= 3 and
                tokens_filtrados[0][0] == 'IDENT' and
                tokens_filtrados[1] == ('OP', '=')):
            nombre_var  = tokens_filtrados[0][1]
            tokens_expr = tokens_filtrados[2:]
            parser      = Parser(tokens_expr)
            nodo_expr   = parser.parsear_expr()
            nodo_ast    = NodoAsignacion(NodoIdentificador(nombre_var), nodo_expr)
        else:
            parser   = Parser(tokens_filtrados)
            nodo_ast = parser.parsear_expr()

        print(f"[PARSER]  {nodo_ast}")

        # FASE 3: Evaluación
        ev = Evaluador(self.entorno)
        if isinstance(nodo_ast, NodoAsignacion):
            valor = ev.evaluar(nodo_ast.expresion)
            self.entorno[nodo_ast.variable.nombre] = valor
            print(f"[EXEC]    {nodo_ast.variable.nombre} = {valor}")
            resultado = valor
        else:
            resultado = ev.evaluar(nodo_ast)
            print(f"[EVAL]    → {resultado}")

        print(f"[ESTADO]  {self.entorno}")
        return resultado

# Ejecutar un pequeño programa
interp = MiniInterprete()
programa = [
    'x = 10',
    'y = 4',
    'x + y',
    'z = x * y',
    'z + 1',
]

print("=" * 55)
print("Ejecutando programa con MiniIntérprete")
print("=" * 55)
for linea in programa:
    interp.ejecutar(linea)
````

**🔍 Análisis:**  
- El intérprete sigue el pipeline completo: **tokenizar → parsear → evaluar** para cada instrucción.  
- El entorno persiste entre instrucciones: `z = x * y` puede acceder a `x` y `y` definidos anteriormente.  
- Este es exactamente el modelo de un REPL de Python, Ruby o JavaScript.

### 📝 Ejercicio 34 — Intérprete con Reporte de Errores

**Concepto:** Un buen intérprete reporta errores con contexto (fase, línea, tipo).  
**Objetivo:** Mejorar el mini-intérprete con manejo robusto de errores.

````python
# EJERCICIO 34 — Reporte de errores con contexto

class ErrorInterprete(Exception):
    def __init__(self, fase, tipo, mensaje, linea=None):
        self.fase    = fase
        self.tipo    = tipo
        self.mensaje = mensaje
        self.linea   = linea
    def __str__(self):
        loc = f" (línea {self.linea})" if self.linea else ""
        return f"[{self.fase}] {self.tipo}{loc}: {self.mensaje}"

def interpretar_con_errores(programa_texto):
    """Intérprete con manejo de errores estructurado."""
    lineas  = programa_texto.strip().split('\n')
    entorno = {}
    ev      = Evaluador(entorno)

    print("=" * 55)
    print("Ejecutando programa:")
    print("=" * 55)

    for num_linea, linea in enumerate(lineas, 1):
        linea = linea.strip()
        if not linea:
            continue

        print(f"\n  Línea {num_linea}: {linea}")

        try:
            # Fase léxica
            try:
                tokens = tokenizar(linea)
                tokens_f = [(t,v) for t,v in tokens
                            if t in ('NUMERO','IDENT','OP','LPAREN','RPAREN')]
            except SyntaxError as e:
                raise ErrorInterprete('LÉXICO', 'SyntaxError', str(e), num_linea)

            # Fase de evaluación simplificada
            try:
                if len(tokens_f) >= 3 and tokens_f[0][0]=='IDENT' and tokens_f[1]==('OP','='):
                    nombre = tokens_f[0][1]
                    parser = Parser(tokens_f[2:])
                    expr   = parser.parsear_expr()
                    valor  = ev.evaluar(expr)
                    entorno[nombre] = valor
                    ev.entorno[nombre] = valor
                    print(f"    ✅ {nombre} = {valor}")
                else:
                    parser    = Parser(tokens_f)
                    expr      = parser.parsear_expr()
                    resultado = ev.evaluar(expr)
                    print(f"    ✅ → {resultado}")
            except NameError as e:
                raise ErrorInterprete('EJECUCIÓN', 'NameError', str(e), num_linea)
            except ZeroDivisionError:
                raise ErrorInterprete('EJECUCIÓN', 'ZeroDivisionError',
                                      'división por cero', num_linea)

        except ErrorInterprete as e:
            print(f"    ❌ {e}")

    print(f"\nEstado final: {entorno}")

# Programa con errores mezclados
prog = """
a = 10
b = a + z
c = 5 / 0
d = a * 2
"""

interpretar_con_errores(prog)
````

**🔍 Análisis:**  
- Los errores se etiquetan por **fase** (léxico, ejecución) y **tipo** (NameError, ZeroDivisionError).  
- El intérprete continúa ejecutando las líneas siguientes tras un error (modo recuperación).  
- `d = a * 2` se ejecuta correctamente porque `a` sí está definida, independientemente de los errores previos.

### 📝 Ejercicio 35 — Intérprete con Funciones Definidas por el Usuario

**Concepto:** El intérprete debe poder almacenar y llamar funciones definidas durante la ejecución.  
**Objetivo:** Extender el intérprete para soportar definición y llamada de funciones.

````python
# EJERCICIO 35 — Intérprete con funciones definidas por el usuario

class InterpreteFunciones:
    """Intérprete que soporta definición y llamada de funciones."""

    def __init__(self):
        self.variables  = {}   # variables globales
        self.funciones  = {}   # {nombre: (params, cuerpo_callable)}

    def definir_funcion(self, nombre, parametros, cuerpo_fn):
        """Registra una función en la tabla de símbolos."""
        self.funciones[nombre] = (parametros, cuerpo_fn)
        print(f"  [DEF] Función '{nombre}' registrada con params={parametros}")

    def llamar_funcion(self, nombre, argumentos):
        """Llama a una función registrada con los argumentos dados."""
        if nombre not in self.funciones:
            raise NameError(f"Función '{nombre}' no definida")

        params, cuerpo = self.funciones[nombre]
        if len(params) != len(argumentos):
            raise TypeError(f"'{nombre}' espera {len(params)} args, recibió {len(argumentos)}")

        # Crear entorno local con los argumentos
        entorno_local = dict(self.variables)     # heredar globales
        for p, a in zip(params, argumentos):
            entorno_local[p] = a

        print(f"  [CALL] {nombre}({dict(zip(params, argumentos))})")
        resultado = cuerpo(**entorno_local)
        print(f"  [RET]  {nombre} → {resultado}")
        return resultado

# Crear intérprete
interp = InterpreteFunciones()
interp.variables['PI'] = 3.14159

# Registrar funciones (cuerpo como lambdas para simplicidad)
interp.definir_funcion('area_circulo',  ['radio'],  lambda radio, **kw: kw.get('PI', 3.14159) * radio ** 2)
interp.definir_funcion('hipotenusa',    ['a', 'b'], lambda a, b, **kw: (a**2 + b**2)**0.5)
interp.definir_funcion('celsius_a_f',   ['c'],      lambda c, **kw: c * 9/5 + 32)

print()
print("Llamadas a funciones:")
print("-" * 45)

r1 = interp.llamar_funcion('area_circulo',  [5])
print(f"  Área círculo r=5: {r1:.4f}\n")

r2 = interp.llamar_funcion('hipotenusa',    [3, 4])
print(f"  Hipotenusa(3,4): {r2:.4f}\n")

r3 = interp.llamar_funcion('celsius_a_f',   [100])
print(f"  100°C en Fahrenheit: {r3}\n")

# Error: función no definida
try:
    interp.llamar_funcion('no_existe', [1])
except NameError as e:
    print(f"  ❌ {e}")
````

**🔍 Análisis:**  
- Las funciones se almacenan en la **tabla de símbolos** junto a las variables (con diferente categoría).  
- Al llamar una función, se crea un entorno **local** que hereda las variables globales pero puede tener sus propias.  
- El intérprete valida el **número de argumentos** antes de ejecutar el cuerpo.

### 📝 Ejercicio 36 — Medición del Overhead de Interpretación

**Concepto:** Los intérpretes tienen overhead por el análisis en tiempo de ejecución; los compiladores no.  
**Objetivo:** Medir y comparar el overhead de interpretación vs ejecución nativa.

````python
# EJERCICIO 36 — Medición del overhead de interpretación
import time

N_ITERACIONES = 10_000

# Método 1: Ejecución nativa (como compilado)
t0 = time.perf_counter()
resultado_nativo = 0
for _ in range(N_ITERACIONES):
    resultado_nativo = (2 + 3) * 4 - 1
t_nativo = time.perf_counter() - t0

# Método 2: Interpretación vía AST
ast_expr = NodoOperacionBinaria('-',
    NodoOperacionBinaria('*',
        NodoOperacionBinaria('+', NodoNumero(2), NodoNumero(3)),
        NodoNumero(4)
    ),
    NodoNumero(1)
)
ev = Evaluador({})

t0 = time.perf_counter()
resultado_ast = 0
for _ in range(N_ITERACIONES):
    resultado_ast = ev.evaluar(ast_expr)
t_ast = time.perf_counter() - t0

# Método 3: Interpretación via eval() de Python (híbrido)
codigo_str = '(2 + 3) * 4 - 1'
t0 = time.perf_counter()
for _ in range(N_ITERACIONES):
    resultado_eval = eval(codigo_str)
t_eval = time.perf_counter() - t0

print(f"Expresión: (2 + 3) * 4 - 1 = {resultado_nativo}")
print(f"Iteraciones: {N_ITERACIONES:,}\n")
print(f"{'Método':<30} {'Tiempo (ms)':<15} {'Overhead relativo'}")
print("-" * 60)
print(f"{'Nativo (Python directo)':<30} {t_nativo*1000:<15.3f} {'1.0x (baseline)'}")
print(f"{'AST Evaluator (nuestro)':<30} {t_ast*1000:<15.3f} {t_ast/t_nativo:.1f}x")
print(f"{'eval() de Python':<30} {t_eval*1000:<15.3f} {t_eval/t_nativo:.1f}x")
print()
print("💡 El overhead de AST se debe a: isinstance() checks,")
print("   llamadas recursivas y acceso a diccionarios en cada operación.")
````

**🔍 Análisis:**  
- El overhead del intérprete AST es visible aunque Python ya es un lenguaje interpretado.  
- En un intérprete C puro vs código máquina, el overhead sería aún más significativo (10x-100x).  
- Esta medición justifica la existencia de los **compiladores JIT** que convierten código interpretado en nativo.

### 📝 Ejercicio 37 — Intérprete con Garbage Collection Básico

**Concepto:** Los intérpretes necesitan gestionar la memoria dinámicamente. El GC libera objetos sin referencias.  
**Objetivo:** Simular un recolector de basura por conteo de referencias.

````python
# EJERCICIO 37 — Simulación de Garbage Collection por conteo de referencias

class ObjetoHeap:
    """Simula un objeto en el heap con conteo de referencias."""
    _id_counter = 0

    def __init__(self, valor):
        ObjetoHeap._id_counter += 1
        self.id          = ObjetoHeap._id_counter
        self.valor       = valor
        self.referencias = 0

    def __repr__(self):
        return f"Obj#{self.id}({self.valor}, refs={self.referencias})"

class GestorHeap:
    """Heap del intérprete con garbage collection por conteo de refs."""

    def __init__(self):
        self._heap = {}   # id → ObjetoHeap

    def asignar(self, valor):
        obj = ObjetoHeap(valor)
        self._heap[obj.id] = obj
        print(f"  [ALLOC] {obj}")
        return obj.id

    def referenciar(self, obj_id):
        self._heap[obj_id].referencias += 1

    def desreferenciar(self, obj_id):
        obj = self._heap.get(obj_id)
        if obj:
            obj.referencias -= 1
            if obj.referencias <= 0:
                print(f"  [GC]    {obj} → referencias=0, liberando")
                del self._heap[obj_id]

    def estado(self):
        print(f"  Heap: {len(self._heap)} objetos")
        for obj in self._heap.values():
            print(f"    {obj}")

# Simulación
heap = GestorHeap()
tabla_vars = {}   # variable → obj_id en heap

print("=== Simulación de Garbage Collection ===\n")

# Asignar objetos
id1 = heap.asignar([1, 2, 3])   # lista
id2 = heap.asignar({'nombre': 'Ana'})
id3 = heap.asignar(42)

# Variables apuntan a objetos
tabla_vars['lista'] = id1;  heap.referenciar(id1)
tabla_vars['datos'] = id2;  heap.referenciar(id2)
tabla_vars['x']     = id3;  heap.referenciar(id3)

print("\nEstado inicial del heap:")
heap.estado()

# Reasignar variable (el objeto anterior pierde una referencia)
print("\nReasignando 'lista' a nuevo valor:")
heap.desreferenciar(tabla_vars['lista'])
id4 = heap.asignar('nuevo_valor')
tabla_vars['lista'] = id4
heap.referenciar(id4)

print("\nEstado del heap después de reasignación:")
heap.estado()

# Salir del ámbito: todas las variables se dereferencian
print("\nSaliendo del ámbito (todas las variables fuera de scope):")
for var, oid in tabla_vars.items():
    heap.desreferenciar(oid)

print("\nEstado final del heap:")
heap.estado()
print("  → Memoria completamente liberada ✅")
````

**🔍 Análisis:**  
- Cuando un objeto llega a 0 referencias, el GC lo libera **automáticamente** (el programador no tiene que hacer `free()`).  
- CPython usa exactamente este mecanismo (refcount) como GC primario.  
- La **reasignación** `lista = nuevo_valor` desreferencia el objeto anterior, pudiendo liberarlo si era la única referencia.

## 🔷 SECCIÓN 8 — Comparación y Sistemas Híbridos

Los sistemas modernos usan arquitecturas **híbridas** que combinan compilación a bytecode con interpretación en VM, más compilación JIT para las partes más ejecutadas.

### 📝 Ejercicio 38 — Simulación de Compilación a Bytecode

**Concepto:** El bytecode es una representación intermedia compacta, más eficiente que el AST para la VM.  
**Objetivo:** Compilar un AST a bytecode y ejecutarlo en una mini-VM.

````python
# EJERCICIO 38 — Compilación AST → Bytecode y ejecución en VM

# Bytecode: lista de (opcode, argumento)
# Opcodes: PUSH_CONST, PUSH_VAR, BINARY_OP, STORE_VAR, PRINT

def ast_a_bytecode(nodo):
    """Compila un nodo AST a lista de instrucciones bytecode."""
    instrucciones = []

    if isinstance(nodo, NodoNumero):
        instrucciones.append(('PUSH_CONST', nodo.valor))

    elif isinstance(nodo, NodoIdentificador):
        instrucciones.append(('PUSH_VAR', nodo.nombre))

    elif isinstance(nodo, NodoOperacionBinaria):
        # Post-orden: compilar hijos primero
        instrucciones += ast_a_bytecode(nodo.izquierda)
        instrucciones += ast_a_bytecode(nodo.derecha)
        instrucciones.append(('BINARY_OP', nodo.operador))

    elif isinstance(nodo, NodoAsignacion):
        instrucciones += ast_a_bytecode(nodo.expresion)
        instrucciones.append(('STORE_VAR', nodo.variable.nombre))

    return instrucciones

def ejecutar_vm(bytecode, entorno=None):
    """Máquina virtual de pila que ejecuta bytecode."""
    if entorno is None:   # evitar el bug del argumento mutable por defecto
        entorno = {}
    pila = []
    ops = {'+': lambda a,b: a+b, '-': lambda a,b: a-b,
           '*': lambda a,b: a*b, '/': lambda a,b: a/b,
           '>': lambda a,b: a>b, '<': lambda a,b: a<b}

    for opcode, arg in bytecode:
        if opcode == 'PUSH_CONST':
            pila.append(arg)
        elif opcode == 'PUSH_VAR':
            pila.append(entorno[arg])
        elif opcode == 'BINARY_OP':
            b = pila.pop()
            a = pila.pop()
            pila.append(ops[arg](a, b))
        elif opcode == 'STORE_VAR':
            entorno[arg] = pila.pop()

    return pila[-1] if pila else None, entorno

# Compilar y ejecutar: z = (x + 10) * 3
entorno_vm = {'x': 5}
ast_vm = NodoAsignacion(
    NodoIdentificador('z'),
    NodoOperacionBinaria('*',
        NodoOperacionBinaria('+', NodoIdentificador('x'), NodoNumero(10)),
        NodoNumero(3)
    )
)

bytecode = ast_a_bytecode(ast_vm)

print("Expresión: z = (x + 10) * 3   con x=5")
print("\nBytecode generado:")
for i, (op, arg) in enumerate(bytecode):
    print(f"  [{i:02d}] {op:<15} {arg}")

print("\nEjecutando en VM de pila:")
_, entorno_resultado = ejecutar_vm(bytecode, entorno_vm)
print(f"\nResultado: z = {entorno_resultado['z']}")
print(f"(esperado: ({entorno_vm['x']} + 10) * 3 = {(entorno_vm['x']+10)*3})")
````

**🔍 Análisis:**  
- El bytecode es más **compacto** que el AST y más fácil de ejecutar: la VM solo hace `push/pop` de una pila.  
- Este es exactamente el modelo de la **JVM** (Java Virtual Machine) y la **CPython VM**.  
- La compilación AST→bytecode ocurre **una sola vez**; luego la VM puede ejecutar el bytecode muchas veces eficientemente.

### 📝 Ejercicio 39 — Simulación de Compilación JIT

**Concepto:** El compilador JIT identifica el código más ejecutado (hot spots) y lo compila a código nativo en tiempo de ejecución.  
**Objetivo:** Simular el comportamiento de un JIT con detección de hot spots.

````python
# EJERCICIO 39 — Simulación de JIT (Just-In-Time compilation)
import time

class MiniJIT:
    """
    Simula un compilador JIT:
    - Las primeras N ejecuciones de una función: INTERPRETA
    - Después del umbral: COMPILA a código nativo (aquí: lambda)
    - Ejecuciones siguientes usan código compilado
    """
    UMBRAL_JIT = 3  # compilar tras 3 invocaciones

    def __init__(self):
        self._contador   = {}   # nombre_func → count
        self._compilado  = {}   # nombre_func → función nativa
        self._funciones  = {}   # nombre_func → (params, ast/fn)

    def registrar(self, nombre, fn_interpretada):
        self._funciones[nombre] = fn_interpretada
        self._contador[nombre]  = 0

    def llamar(self, nombre, *args):
        self._contador[nombre] += 1
        count = self._contador[nombre]

        if nombre in self._compilado:
            # RUTA COMPILADA (JIT activo)
            t = time.perf_counter()
            resultado = self._compilado[nombre](*args)
            elapsed = time.perf_counter() - t
            print(f"  [{count:2d}] {nombre}({args}) = {resultado} "
                  f"[⚡ JIT compilado, {elapsed*1e6:.2f} µs]")
        else:
            # RUTA INTERPRETADA
            t = time.perf_counter()
            resultado = self._funciones[nombre](*args)
            elapsed = time.perf_counter() - t
            print(f"  [{count:2d}] {nombre}({args}) = {resultado} "
                  f"[🐢 Interpretado,  {elapsed*1e6:.2f} µs]")

            # ¿Alcanzamos el umbral?
            if count >= self.UMBRAL_JIT:
                print(f"       *** HOT SPOT detectado → compilando JIT ***")
                # Simular compilación: usar la misma función (ya es Python nativo)
                self._compilado[nombre] = self._funciones[nombre]

        return resultado

# Función a optimizar
def calcular_triangular_interpretado(n):
    """n * (n+1) / 2 — número triangular"""
    return n * (n + 1) // 2

jit = MiniJIT()
jit.registrar('triangular', calcular_triangular_interpretado)

print("Ejecución con JIT (umbral = 3 llamadas):")
print("-" * 60)
for i in range(1, 9):
    jit.llamar('triangular', i * 5)

print(f"\nTotal llamadas: {jit._contador['triangular']}")
print(f"JIT activado después de llamada #{MiniJIT.UMBRAL_JIT}")
````

**🔍 Análisis:**  
- Las primeras 3 invocaciones se **interpretan** (costosas por el overhead de análisis).  
- A partir de la 4ª, el código ya está **"compilado" JIT** y se ejecuta más rápido.  
- En la JVM real, el JIT hace análisis de código, optimizaciones (inlining, escape analysis) y genera código máquina nativo: mucho más impactante que esta simulación.
